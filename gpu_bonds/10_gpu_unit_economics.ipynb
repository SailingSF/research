# 10 · What does one GPU earn? Rent, resale value, payback, and margin vs. the cost of debt

**The question.** A reader (a bottom-up investor) asked for the per-unit view: *"A company that borrows money at 10% but has 30% operating margins is a very different risk profile than one with 12% margins."* Their back-of-envelope: a used A100 costs ~$6k, rents for ~$1.50/hr, costs up to ~$0.50/hr in power and space, so at 90% utilization it pays for itself in about a year. This notebook checks that arithmetic for each GPU generation, then asks what it means for CoreWeave's bondholders.

**Short answer** (prices as of Nov 9 2026; CoreWeave Q2 2026 filings):
1. **GPU rent stopped falling in 2026** (§1, `10_rent_history.png`). The H100 fell ~80% from its 2023 peak to a low of ~$1.50–2.00 an hour in late 2025 / early 2026, then rose: ~$2.50 spot and ~$3.10–3.70 in executed trades and list prices now. Measured from Jan 2024, the H100 is roughly flat to up, not the ~19% a year 08 §3 used. **This revises 08 §3**, though one good year doesn't make a trend. Newer chips rent for more: B200 ~$9.40 an hour executed.
2. **Used GPUs hold their value better than the "falling rents" story implies** (§2, `10_new_vs_used.png`). Broker quotes put a used H100 at ~$15–28k against ~$25–40k new: ~75% of a new card three years in. But used-price sources disagree by up to 2×, and a used card alone is ~60% of what a lender financed (server, networking and installation don't resell as well).
3. **The reader's arithmetic checks out for one GPU at today's prices** (§3, `10_payback.png`). Their A100 pays back in ~0.8 years. Today's running cost (power + wholesale data-center space) is ~$0.43/hr for an H100, so their $0.50 is about right. A new H100 system bought today pays back in ~1.8 years at 90% utilization; a B200 in under a year. **CoreWeave as a whole takes ~5.5 years.** The gap: CoreWeave sells multi-year contracts, not spot hours, and its average revenue works out to roughly $1.35–2.10 per GPU-hour, below today's spot prices.
4. **CoreWeave is the reader's "borrows at 10%" case, with a ~0% margin, not 12%** (§4, `10_margins.png`). Q2 operating margin was −2% (59% before depreciation). Per $1 of equipment, the business earns ~18¢ of cash a year, ~17¢ of which is the GPU wearing out over 6 years, leaving ~1.5¢ against ~6.4¢ of interest. Equinix and Digital Realty, the reader's long-run destination, run 16–23% operating margins but borrow at ~3% for 20-year buildings. At Equinix's margin, CoreWeave's equipment would earn ~7% a year against a 9.5% borrowing cost.
5. **A second life helps a lot, but only gets CoreWeave to about its cost of debt** (§5, `10_second_life.png`). Most of the −2% margin is depreciation over 6 years. A ~6½-year-old A100 still earns ~30¢ a year per dollar it cost, the same as CoreWeave's new equipment under contract. If CoreWeave's chips do the same after their contracts end, its equipment returns ~9.4% a year over an 8-year life at 90% utilization (~5% at 60% utilization or at the 2026 glut price), against 9.5% interest. Lenders get repaid by about year 6. If the chips die with their contracts, they never do.
6. **CoreWeave's contracts are cheap because they were signed near the bottom, at prices set to cover cost, in exchange for financing** (§6, `10_contracts.png`).
   - ~$64B of the ~$89B of contracts with known amounts were signed Mar 2025–Apr 2026, while H100 prices fell to their low. Meta's $21B was signed the month prices bottomed.
   - CoreWeave earns ~$9M per MW of power a year. Its new 3–6-month contracts are priced at ~$40M per MW. Today's executed B200 price would match its revenue at ~20% utilization.
   - It can't simply switch to hourly rental. The hourly market is thin, and the take-or-pay contracts are what lenders lent against.
   - Meta's contracts are fixed to 2031–32, so repricing that equipment waits until then.
   - For shareholders, the payoff is the second life. The loan is repaid within the contract, so a 10-year life at today's A100-like rents returns ~15% a year on their stake, against −48% if the chips die with the contracts.
7. **CoreWeave's chips should outlive their contracts by ~2½ years, worth ~$9B today, ~33% of the stock's value** (§7, `10_chip_lives.png`).
   - ~¾ of its 1.5 GW is Blackwell installed in the last 18 months (our estimate; CoreWeave doesn't disclose the split).
   - Azure and Google retired past generations 7–9.5 years after launch (median 8.3), though V100s are still listed by 17 providers at 9.4 years.
   - Because CoreWeave installs 0.5–1.5 years after launch and sells ~5-year contracts, the median life leaves ~2.5 years after each contract, not 4–5. A second life pays shareholders well, because the loans are gone. But a 10-year-from-launch life is worth ~$14B, still under the ~$27B market value. The rest of the stock's value has to come from equipment not yet built.
8. **After Q3, the new contracts make money while they run, but the old fleet's second life is largely promised to the holding company's bondholders** (§8, `10_q3_leverage.png`).
   - New 4–5-year contracts at $20–25M per MW cover their loan payments ~3× (the old book: 1.2×) and repay the shareholders' stake within about a year. Lenders cap loans at 70–100% of cost, so higher prices mean faster repayment, not more debt.
   - The exception is DDTL 5.5 ($2.6B): 3-year contracts against 5-year debt. In a glut, years 4–5 fall a few $M per MW short, small next to what those contracts pay.
   - CoreWeave, Inc., the listed holding company, owes ~$21B in its own bonds, due in single payments in 2030–33, about what the old fleet's second life earns at the base price in those years.
   - Management's renewal plan (~$10B a year more from 2027–28 roll-offs) implies ~$5.15 per GPU-hour. That's today's shortage pricing, lasting two more years. If it does, the second life is several times the bonds.
   - Revises §7: valued per MW, the second life of today's fleet is worth ~$16B (base), ~60% of the stock's ~$27B.
9. **The old contracts can't pay CoreWeave, Inc.'s own bonds; the new contracts and the GPUs' second life can, even in a glut** (§9, `10_escape.png`).
   - On the old book alone (~11¢ of each revenue dollar left after the GPU loans), CoreWeave, Inc. runs out of cash in 2031 against its ~$21B of bonds due 2030–33.
   - Q3's new contracts (~58¢ left per dollar) add ~$19B, and the second life ~$27B. If CoreWeave signed nothing more, it would end 2033 with ~$29B (~$16B at glut prices).
   - The risk is the next round: ~$38B of capex in 2027, priced at whatever the market pays then.

```python
import json
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import matplotlib.dates as mdates
from common import INK, INK2, MUTED, AXIS, GRID, BOND, RAMP, EQUITY, source_note

plt.rcParams['text.parse_math'] = False
CHARTS = Path('charts')
INP = pd.read_csv('data/neocloud_inputs.csv').set_index('item')
v = lambda k: float(INP.loc[k, 'value'])
U = pd.read_csv('data/gpu_units.csv')
HIST = pd.read_csv('data/gpu_rent_history.csv', parse_dates=['date'])
GD = pd.read_csv('data/gpu_prices/getdeploying_weekly.csv', parse_dates=['date'])
GENS = ['A100', 'H100', 'H200', 'B200']
HOURS = 8760
OUT = {}   # headline numbers for data/gpu_economics.csv

def ornn(name):
    d = pd.DataFrame(json.load(open(f'data/gpu_prices/ornn_{name}.json'))['data'])
    return pd.Series(d.index_value.values, pd.to_datetime(d.timestamp).dt.tz_localize(None).dt.normalize())

def gd(gpu, billing):
    x = GD[(GD.gpu_slug == f'nvidia-{gpu.lower()}') & (GD.billing_type == billing)]
    return x.set_index('date').provider_median_price.sort_index()

def mid(gpu, measure):
    r = U[(U.gpu == gpu) & (U.measure == measure)]
    return r[['low', 'high']].values.mean() if len(r) else np.nan

ORNN = {'A100': ornn('A100_SXM4'), 'H100': ornn('H100_SXM'), 'H200': ornn('H200'), 'B200': ornn('B200')}
```

## 1 · What does a GPU rent for, and is the price still falling?

**Why ask.** Every number downstream (payback, margin, what an old GPU earns after its contract) is a rental price times hours sold. 08 §3 assumed H100 rents keep falling ~19% a year, the Jan 2024–Oct 2025 pace. If that's wrong, so is the size of the bondholders' problem.

**Method.** There's no single official GPU price, so we show four measures and say which is which:
- **Marketplace median** (Silicon Data, via the SpaceX post): what small renters paid on open marketplaces, 2023–2025.
- **Hyperscaler list** (Silicon Data): AWS/Azure/Google sticker prices. Big customers don't pay these.
- **List prices, weekly** (GetDeploying): the median across ~40 providers, on-demand and spot, since Nov 2025.
- **Executed trades, daily** (Ornn Compute Price Index): volume-weighted prices of actual neocloud rental deals, Aug–Nov 2026. This is the closest thing to a market price, but it covers short-term deals, not 5-year contracts.

The 30-day median of Ornn's index is our "today's price" in §3, because daily prints jump around (H200 rose 19% in the last month alone).

```python
mkt = HIST[HIST.series == 'h100_marketplace'].set_index('date').usd_per_hr
hyp = HIST[HIST.series == 'h100_hyperscaler'].set_index('date').usd_per_hr
od, spot = gd('H100', 'ON_DEMAND'), gd('H100', 'SPOT')

rent_now = pd.DataFrame({g: {'executed, 30-day median (Ornn)': ORNN[g].iloc[-30:].median(),
                             'on-demand list (GetDeploying)': gd(g, 'ON_DEMAND').iloc[-1],
                             'spot list (GetDeploying)': gd(g, 'SPOT').iloc[-1]} for g in GENS}).T
NOW = rent_now['executed, 30-day median (Ornn)']

yrs = lambda a, b: (b - a).days / 365.25
peak, jan24 = mkt.loc['2023-06-01'], mkt.loc['2024-01-01']
low = min(mkt.min(), spot.min())
cagr = lambda p0, p1, t: (p1 / p0) ** (1 / t) - 1
trend = pd.Series({
    '08 §3: Jan 2024 → Oct 2025 (marketplace)': cagr(jan24, mkt.loc['2025-10-01'], yrs(pd.Timestamp('2024-01-01'), pd.Timestamp('2025-10-01'))),
    'Jan 2024 (multi-provider avg) → now (on-demand list)': cagr(jan24, od.iloc[-1], yrs(pd.Timestamp('2024-01-01'), od.index[-1])),
    'Jan 2024 → now (spot list)': cagr(jan24, spot.iloc[-1], yrs(pd.Timestamp('2024-01-01'), spot.index[-1])),
    'Nov 2025 → now (on-demand list)': cagr(od.iloc[0], od.iloc[-1], yrs(od.index[0], od.index[-1])),
    'Aug → Nov 2026 (executed, Ornn)': ORNN['H100'].iloc[-1] / ORNN['H100'].iloc[0] - 1}, name='change a year (last row: 3 months, not annualized)')
OUT.update({'h100_peak_2023': peak, 'h100_low': low, 'h100_now_executed': NOW['H100'], 'h100_now_spot': spot.iloc[-1],
            'h100_now_ondemand': od.iloc[-1], 'h100_change_jan24_to_now_spot': trend.iloc[2],
            'h100_change_jan24_to_now_ondemand': trend.iloc[1], 'h100_fall_from_peak_to_low': 1 - low / peak})
print(f'H100: ${peak:.2f} peak (2023) → ${low:.2f} low → ${NOW["H100"]:.2f} executed / ${spot.iloc[-1]:.2f} spot now. Fall peak→low: {1 - low / peak:.0%}')
display(trend.map('{:+.0%}'.format).to_frame(), rent_now.round(2))
```

```python
fig, (ax, bx) = plt.subplots(1, 2, figsize=(11.5, 4.8), gridspec_kw={'width_ratios': [2.3, 1]})
ax.plot(hyp.index, hyp, color=AXIS, lw=1.6, marker='o', ms=3.5, label='Hyperscaler list price (AWS/Azure/Google sticker)')
ax.plot(mkt.index, mkt, color=RAMP[2], lw=2, marker='o', ms=3.5, label='Marketplace median (Silicon Data)')
ax.plot(od.index, od, color=RAMP[1], lw=1.6, label='On-demand list, median of ~40 providers (GetDeploying)')
ax.plot(spot.index, spot, color=RAMP[0], lw=1.6, label='Spot list (GetDeploying)')
ax.plot(ORNN['H100'].index, ORNN['H100'], color=EQUITY, lw=1.4, label='Executed trades (Ornn index, daily)')
lo_date = spot.idxmin()
ax.annotate(f'Low: ~${spot.min():.2f} spot\n(early 2026)', (lo_date, spot.min()), xytext=(lo_date - pd.Timedelta(days=330), 0.6),
            fontsize=8.5, color=INK, arrowprops=dict(arrowstyle='-', color=MUTED, lw=0.8))
ax.annotate(f'2023 shortage:\n~${peak:.0f}/hr', (mkt.index[0], peak), xytext=(mkt.index[0] + pd.Timedelta(days=60), 9.6), fontsize=8.5, color=INK,
            arrowprops=dict(arrowstyle='-', color=MUTED, lw=0.8))
ax.set_ylim(0, 11); ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f'${y:.0f}'))
ax.xaxis.set_major_locator(mdates.YearLocator()); ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax.set_ylabel('Rent, $ per GPU-hour'); ax.grid(axis='x', visible=False)
ax.legend(loc='upper right', fontsize=8)
ax.set_title('The H100: down ~80% to early 2026, then up', fontsize=11)

cols = [(rent_now.columns[2], RAMP[0]), (rent_now.columns[0], EQUITY), (rent_now.columns[1], RAMP[1])]
w = 0.26
for j, (c, colr) in enumerate(cols):
    bx.barh(np.arange(len(GENS)) + (j - 1) * w, rent_now[c], w, color=colr, label=c.split(' (')[0])
for i, g in enumerate(GENS):
    bx.text(rent_now.loc[g].max() + 0.2, i, f'${NOW[g]:.2f}', va='center', fontsize=8.5, color=EQUITY, fontweight='bold')
bx.set_yticks(range(len(GENS)), GENS); bx.invert_yaxis(); bx.grid(axis='y', visible=False)
bx.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:.0f}')); bx.set_xlim(0, 12)
bx.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=3, fontsize=7.5, handlelength=1)
bx.set_title(f'Today, by generation ({ORNN["H100"].index[-1]:%b %-d %Y})', fontsize=11)
fig.suptitle('What a GPU rents for', x=0.01, ha='left', fontweight='bold', fontsize=13)
source_note(fig, 'Source: Silicon Data via the SpaceX post (marketplace and hyperscaler, 2023–25); GetDeploying weekly provider medians; '
                 'Ornn Compute Price Index (executed neocloud trades; bold label = 30-day median). Methods differ; read levels across lines with care.')
fig.tight_layout(rect=(0, 0.03, 1, 1))
fig.savefig(CHARTS / '10_rent_history.png', dpi=200, bbox_inches='tight')
plt.show()
```

**What this shows** (`charts/10_rent_history.png`, above):
- **The collapse was real, and it ended.** The H100 went from ~$8 an hour in the 2023 shortage to ~$2 on marketplaces by mid-2025, and spot list prices bottomed around $1.50 in Feb–Apr 2026. Since then, every measure has risen: spot ~$2.50, executed trades ~$3.10, on-demand list ~$3.70.
- **Measured from Jan 2024, the H100 is roughly flat.** Spot is down ~5% a year; on-demand list is up ~10% a year. 08 §3 used ~19% a year, a window (Jan 2024–Oct 2025) that ended at the bottom. **This revises 08 §3:** lenders there are whole if old-GPU prices fall no faster than ~14% a year, and since Jan 2024 they have. That's a real point for the bonds. It's also one year of a shortage (the same tightness that sent B200 executed prices up ~70% since August), and a three-year-old chip renting for more than it did two years ago is what a shortage looks like, not a law.
- **Hyperscaler sticker prices barely matter.** They sit at $6–9 and fell only after AWS's 2025 price cut. Big customers negotiate far below them.
- **Newer chips rent for more, roughly in line with what they cost** (§2): B200 ~$9.40 executed, H200 ~$6.30, H100 ~$3.10, A100 ~$1.05. The reader's $1.50 for an A100 sits between executed trades and spot lists.

**Next:** if rents held up, did resale values?

## 2 · What is a GPU worth new vs. used?

**Why ask.** For a lender, a GPU's resale price is the recovery value if the borrower fails: what the collateral fetches in a forced sale. For the reader's math, it's the purchase price.

**Method.** Dealer and broker price ranges for each generation, new and used, from `data/gpu_units.csv`. These are indicative quotes, not transactions, and **the sources disagree by up to 2×** (e.g. a used A100 80GB: $7,000–9,500 from one broker index, $15,000–17,000 from another). We show each source's range separately rather than averaging them away. For scale, each row also marks **one year of rent** at today's executed price and 90% utilization: how much of the card's price a single year earns.

Lenders didn't finance a bare card. They financed a server and networking around it, which we put at the 8-GPU server price ÷ 8, plus ~$2,500 of networking per GPU.

```python
px = U[U.gpu.isin(GENS) & U.measure.isin(['new_gpu', 'used_gpu', 'server_per_gpu'])].copy()
px['src'] = px.source_url.str.extract(r'https?://(?:www\.)?([^/]+)')[0]
year_rent = NOW * 0.9 * HOURS
allin = pd.Series({g: mid(g, 'server_per_gpu') + mid('all', 'network_per_gpu') for g in GENS})
used_mid = px[px.measure == 'used_gpu'].groupby('gpu')[['low', 'high']].apply(lambda d: d.values.mean())
new_mid = px[px.measure == 'new_gpu'].groupby('gpu')[['low', 'high']].apply(lambda d: d.values.mean())
tab = pd.DataFrame({'new card (mid)': new_mid, 'used card (mid)': used_mid, 'used ÷ new': used_mid / new_mid,
                    'all-in per GPU (server + network)': allin, 'used card ÷ all-in': used_mid / allin,
                    'one year of rent (90%)': year_rent}).reindex(GENS)
OUT.update({'h100_used_over_new': tab.loc['H100', 'used ÷ new'], 'h100_used_over_allin': tab.loc['H100', 'used card ÷ all-in']})
tab.style.format({c: '${:,.0f}' for c in tab.columns if '÷' not in c} | {c: '{:.0%}' for c in tab.columns if '÷' in c}, na_rep='—')
```

```python
fig, ax = plt.subplots(figsize=(10.5, 5.2))
rows, y = [], 0
style = {'new_gpu': (RAMP[2], 'New card'), 'used_gpu': (RAMP[0], 'Used card'), 'server_per_gpu': (AXIS, 'Server + share of system, per GPU')}
seen = set()
for g in GENS:
    for meas in ['new_gpu', 'used_gpu', 'server_per_gpu']:
        for _, r in px[(px.gpu == g) & (px.measure == meas)].iterrows():
            colr, lab = style[meas]
            ax.plot([r.low, max(r.high, r.low + 400)], [y, y], color=colr, lw=7, solid_capstyle='butt', label=lab if lab not in seen else None)
            seen.add(lab)
            ax.text(r.high + 700, y, r.src, va='center', fontsize=7.5, color=MUTED)
            rows.append((y, g)); y += 1
    ax.scatter([year_rent[g]], [y - 1 - 0.5 * (len(px[px.gpu == g]) - 1)], marker='D', s=46, color=EQUITY, zorder=4,
               label='One year of rent at today\'s executed price, 90% utilized' if g == 'A100' else None)
    ax.axhline(y - 0.5 + 0.0, color=GRID, lw=0.8)
    y += 0.6
ticks = pd.DataFrame(rows, columns=['y', 'g']).groupby('g', sort=False).y.mean()
ax.set_yticks(ticks.values, ticks.index, fontsize=10, fontweight='bold'); ax.invert_yaxis(); ax.grid(axis='y', visible=False)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x / 1e3:.0f}k')); ax.set_xlim(0, 80000)
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.12), ncol=2, fontsize=8.5)
ax.set_xlabel('Price per GPU')
ax.set_title('A used H100 fetches ~75% of a new card; a year of rent covers most of a used card\n', pad=10)
ax.text(0, 1.02, 'No used B200s trade yet. Each bar is one dealer or broker source; they disagree by up to 2×.', transform=ax.transAxes, fontsize=8.5, color=INK2)
source_note(fig, 'Source: data/gpu_units.csv (compute.exchange Q3 2026, serverbuyback.com Jun 2026, intuitionlabs.ai Sep 2026, DGX A100 launch price). '
                 'Indicative quotes, not transactions. Rent: Ornn index 30-day median, Nov 2026.')
fig.tight_layout(rect=(0, 0.03, 1, 1))
fig.savefig(CHARTS / '10_new_vs_used.png', dpi=200, bbox_inches='tight')
plt.show()
```

**What this shows** (`charts/10_new_vs_used.png` and the table above):
- **Used prices say the market expects these chips to keep earning.** A used H100 is quoted at ~75% of a new card three years after launch (averaging both sources); a used H200 at roughly what a new one costs on the street. Resale tracks rent, and rent rose in 2026 (§1).
- **The A100 is the reader's case.** Their ~$6k used price is the A100 40GB range ($4–6k); the 80GB is $7–17k depending on the source. At today's executed price (~$1.05) one year of rent is ~$8,300, so a used A100 card earns its price back in about a year, before costs. That's the reader's point, and it holds.
- **For a lender, the card isn't the collateral.** A lender financed the whole system: ~$38k per GPU for an H100 (server + networking). A used H100 card (~$23k, both sources' midpoint) recovers ~60% of that, and the rest of the server resells for less. Add forced-sale discounts and dealers' margins, and recovery on GPU collateral is well below the card quotes here.
- **The sources are noisy.** Two broker indexes disagree by 2× on the A100 80GB. Treat these as ranges, not prices.

**Next:** with rent (§1) and price (§2), how long does one GPU take to pay for itself?

## 3 · How fast does one GPU pay for itself, and why is CoreWeave so much slower?

**Why ask.** This is the reader's question directly: is renting GPUs a 1-year-payback business or a 5-year one? Payback is what the bond lender cares about, because a loan that's paid off before the chip goes stale is a safe loan.

**Method.** For one GPU:

$$\text{payback (years)} = \frac{\text{what it cost}}{(\text{rent per hour} \times \text{utilization} - \text{running cost per hour}) \times 8{,}760}$$

- **Running cost per hour** = the GPU's power draw (kW, from Nvidia's system specs) × (power at 8¢/kWh × 1.3 for cooling + wholesale data-center rent). Data-center rent is CBRE's H1 2026 North American asking rate for large deployments, ~$170 per kW a month, about 23¢ per kW-hour. For an H100 this comes to ~$0.43/hr, close to the reader's $0.50. It leaves out staff, sales and overhead (~9% of CoreWeave's revenue).
- **Utilization:** the reader's 90%, and 60% as a check (the SpaceX post used 65%).
- **Cases:**
  - the reader's A100, card only, at their numbers;
  - a new H100, H200 or B200 system bought today, at today's executed rent (§1);
  - an H100 system bought in mid-2023 at ~$42,500 all-in (the top of today's server range plus networking; 2023 prices were higher), rented at the market price every month since (§1's marketplace line, then spot);
  - **CoreWeave, company average:** 1 ÷ the cash it earns a year per $1 of equipment (Adjusted EBITDA ÷ equipment cost, the 18¢ from 08 §3).

```python
colo_hr = mid('all', 'colo_usd_kw_month') / 730
RUN = pd.Series({g: mid(g, 'kw_per_gpu') * (v('pue') * v('power_price') + colo_hr) for g in GENS})

def payback(cost, rate, util, run):
    net = (rate * util - run) * HOURS
    return cost / net if net > 0 else np.inf

# H100 bought Jun 2023, rented at the market price each month since (marketplace to Oct 2025, then spot)
path = pd.concat([mkt, spot.resample('MS').mean().loc['2025-11':]]).resample('MS').mean().interpolate()
def months_to_payback(cost, util):
    cum = ((path * util - RUN['H100']) * HOURS / 12).cumsum()
    return (np.argmax(cum.values >= cost) + 1) / 12 if (cum >= cost).any() else np.inf
COST_2023 = U[(U.gpu == 'H100') & (U.measure == 'server_per_gpu')].high.iloc[0] + mid('all', 'network_per_gpu')

fl = pd.read_csv('data/fleet.csv').query("issuer == 'CoreWeave'").gross_bn.sum()
rev_gw = v('crwv_rev_q2') * 4 / ((v('crwv_gw_start_q2') + v('crwv_gw_end_q2')) / 2)
cost_gw = fl / v('crwv_gw_end_q2')
MARGIN = v('crwv_adj_ebitda_q2') / v('crwv_rev_q2')
CASH_PER_DOLLAR = MARGIN * rev_gw / cost_gw          # 08 §3's ~18¢
REV_PER_DOLLAR = rev_gw / cost_gw

cases = {
    "Reader's A100: used card, $6k, $1.50/hr, $0.50/hr costs": (payback(6000, 1.5, 0.9, 0.5), payback(6000, 1.5, 0.6, 0.5)),
    **{f'New {g} system today, ${allin[g] / 1e3:.0f}k, ${NOW[g]:.2f}/hr': (payback(allin[g], NOW[g], 0.9, RUN[g]), payback(allin[g], NOW[g], 0.6, RUN[g]))
       for g in ['H100', 'H200', 'B200']},
    f'H100 bought mid-2023, ${COST_2023 / 1e3:.1f}k, market rent since': (months_to_payback(COST_2023, 0.9), months_to_payback(COST_2023, 0.6)),
    'CoreWeave, company average (Q2 2026)': (1 / CASH_PER_DOLLAR, np.nan)}
PB = pd.DataFrame(cases, index=['90% utilized', '60% utilized']).T
# what CoreWeave's revenue works out to per GPU-hour, if every GPU billed every hour, for H100- and B200-class systems
implied = pd.Series({g: REV_PER_DOLLAR * allin[g] / HOURS for g in ['H100', 'B200']})
OUT.update({'payback_reader_a100': PB.iloc[0, 0], 'payback_h100_new_today': PB.iloc[1, 0], 'payback_b200_new_today': PB.iloc[3, 0],
            'payback_h100_2023': PB.iloc[4, 0], 'payback_crwv_avg': PB.iloc[5, 0], 'run_cost_h100_hr': RUN['H100'],
            'crwv_rev_per_gpu_hr_h100_class': implied['H100'], 'crwv_rev_per_gpu_hr_b200_class': implied['B200']})
print('Running cost per hour:', RUN.round(2).to_dict())
print(f"CoreWeave revenue per $1 of equipment: {REV_PER_DOLLAR:.2f}/yr; cash {CASH_PER_DOLLAR:.2f}/yr. "
      f"As $/GPU-hour at 100% billing: H100-class ${implied['H100']:.2f}, B200-class ${implied['B200']:.2f}")
PB.round(1)
```

## 4 · Is the operating margin above the cost of the debt?

**Why ask.** The reader's test: borrowing at 10% with a 30% operating margin is a different risk from borrowing at 10% with 12%. Operating margin is profit after the equipment wears out (depreciation) and before interest. The reader also expects GPU margins to drift toward data-center landlords like Equinix (EQIX) and Digital Realty (DLR), so we put those beside CoreWeave.

**Method.**
- **Per $100 of revenue,** from each company's latest 10-Q (SEC XBRL): depreciation, other operating costs, operating profit, and interest. CoreWeave: Q2 2026. Equinix: Q3 2026. Digital Realty: Q1 2026 (its latest quarter in the SEC's data).
- **The comparison that matters for a lender** is per $1 of assets, not per $1 of revenue: operating profit ÷ total assets vs. interest ÷ debt. A 20% margin on a business that turns over 15¢ of revenue per $1 of assets earns 3% on its assets.
- **CoreWeave per $1 of equipment:** cash a year (18¢, §3), minus wear-out (1/6 a year over its 6-year accounting life), against interest (68¢ borrowed per $1 × 9.5%, 08 §3).
- Margins are annualized from one quarter, so seasonality and one-offs aren't smoothed.

```python
def facts(tk):
    f = json.load(open(f'data/edgar/companyfacts_{tk}.json'))['facts']['us-gaap']
    def get(tags, end=None, start=None, instant=False):
        for t in tags:
            if t not in f: continue
            u = next(iter(f[t]['units'].values()))
            for x in u:
                if x['end'] == end and (instant and 'start' not in x or x.get('start') == start):
                    return x['val'] / 1e9
        return np.nan
    return get

REV = ['Revenues', 'RevenueFromContractWithCustomerExcludingAssessedTax']
DA = ['DepreciationDepletionAndAmortization', 'DepreciationAndAmortization']
INT = ['InterestExpenseNonoperating', 'InterestExpense', 'InterestExpenseDebt']
DEBT = {'CRWV': ['DebtInstrumentCarryingAmount'], 'EQIX': ['LongTermDebt'], 'DLR': ['SeniorNotes', 'SecuredDebt', 'UnsecuredDebt']}
Q = {'CRWV': ('2026-04-01', '2026-06-30', '2026-03-31'), 'EQIX': ('2026-07-01', '2026-09-30', '2026-06-30'),
     'DLR': ('2026-01-01', '2026-03-31', '2025-12-31')}
NAME = {'CRWV': 'CoreWeave (Q2 2026)', 'EQIX': 'Equinix (Q3 2026)', 'DLR': 'Digital Realty (Q1 2026)'}
rows = {}
for tk, (s, e, prev) in Q.items():
    g = facts(tk)
    rev, op = g(REV, e, s), g(['OperatingIncomeLoss'], e, s)
    da = g(DA, e, s) if tk != 'CRWV' else g(DA, e, '2026-01-01') - g(DA, prev, '2026-01-01')   # CoreWeave tags D&A year-to-date only: H1 − Q1
    intr = g(INT, e, s) if tk != 'CRWV' else v('crwv_interest_q2')
    assets = (g(['Assets'], e, instant=True) + g(['Assets'], prev, instant=True)) / 2
    debt = sum(np.nan_to_num(g([t], e, instant=True)) for t in DEBT[tk])
    rows[NAME[tk]] = {'revenue ($bn, qtr)': rev, 'depreciation': da / rev, 'operating margin': op / rev, 'interest': intr / rev,
                      'other operating costs': 1 - op / rev - da / rev,
                      'operating profit ÷ assets': 4 * op / assets, 'interest ÷ debt': 4 * intr / debt}
M = pd.DataFrame(rows).T
M.loc['CoreWeave (Q2 2026)', 'interest ÷ debt'] = (v('crwv_interest_q2') + v('crwv_capint_q2')) * 4 / ((v('crwv_debt_q1') + v('crwv_debt_q2')) / 2)   # 08 §3's RATE
LOAN = v('crwv_debt_q2') / v('crwv_ppe_gross')
RATE = M.loc['CoreWeave (Q2 2026)', 'interest ÷ debt']
per_dollar = pd.Series({'cash earned (Adjusted EBITDA)': CASH_PER_DOLLAR, 'wear-out (1/6 a year)': 1 / 6,
                        'left after wear-out': CASH_PER_DOLLAR - 1 / 6, f'interest ({LOAN:.0%} borrowed × {RATE:.1%})': LOAN * RATE})
eqix_like = M.loc['Equinix (Q3 2026)', 'operating margin'] * REV_PER_DOLLAR
OUT.update({'crwv_op_margin': M.iloc[0]['operating margin'], 'crwv_interest_share_rev': M.iloc[0]['interest'],
            'eqix_op_margin': M.iloc[1]['operating margin'], 'dlr_op_margin': M.iloc[2]['operating margin'],
            'crwv_left_after_wearout': per_dollar.iloc[2], 'crwv_interest_per_dollar': per_dollar.iloc[3],
            'crwv_return_at_eqix_margin': eqix_like, 'crwv_rate': RATE})
print(f"At Equinix's operating margin, CoreWeave's equipment would earn {eqix_like:.1%} a year against {RATE:.1%} interest.")
display(M.style.format({'revenue ($bn, qtr)': '{:.2f}'} | {c: '{:.1%}' for c in M.columns[1:]}), per_dollar.map('{:.3f}'.format).to_frame('per $1 of equipment, a year'))
```

```python
fig, (ax, bx) = plt.subplots(1, 2, figsize=(11.5, 4.4), gridspec_kw={'width_ratios': [1.5, 1]})
parts = [('other operating costs', AXIS, 'Other operating costs'), ('depreciation', RAMP[0], 'Depreciation (wear-out)'),
         ('operating margin', RAMP[2], 'Operating profit')]
for i, co in enumerate(M.index):
    left = 0
    for key, colr, lab in parts:
        w = M.loc[co, key] * 100
        if w > 0: ax.barh(i, w, left=left, color=colr, height=0.5, label=lab if i == 0 else None)
        if abs(w) > 6:
            ax.text(left + w / 2, i, f'{w:.0f}', ha='center', va='center', fontsize=8.5, color='white' if colr != AXIS else INK, fontweight='bold')
        left += max(w, 0)
    op = M.loc[co, 'operating margin'] * 100
    ax.text(105, i - 0.05, f'Operating margin {op:+.0f}%\nInterest {M.loc[co, "interest"] * 100:.0f}% of revenue', fontsize=8, color=INK, va='center')
ax.set_yticks(range(len(M)), M.index, fontsize=9); ax.invert_yaxis(); ax.grid(axis='y', visible=False)
ax.set_xlim(0, 104); ax.set_xlabel('Per $100 of revenue (CoreWeave\'s costs add to $102: a $2 loss)')
ax.plot([0, 0], [0, 0], color=RAMP[2], lw=6, label='Operating profit')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.17), ncol=3, fontsize=8)
ax.set_title('Where each $100 of revenue goes', fontsize=11)

lab = ['Cash earned', 'GPU wear-out', 'Left over', 'Interest owed']
vals = per_dollar.values * 100
cols = [RAMP[1], RAMP[0], RAMP[2], EQUITY]
bx.bar(range(4), vals, color=cols, width=0.6)
for i, val in enumerate(vals):
    bx.text(i, val + 0.4, f'{val:.1f}¢', ha='center', fontsize=9.5, fontweight='bold', color=INK)
bx.set_xticks(range(4), lab, fontsize=8.5); bx.grid(axis='x', visible=False)
bx.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f'{y:.0f}¢')); bx.set_ylim(0, 22)
bx.set_title('CoreWeave, per $1 of equipment a year', fontsize=11)
fig.suptitle('Borrowing at ~9.5% on a ~0% operating margin', x=0.01, ha='left', fontweight='bold', fontsize=13)
source_note(fig, 'Source: SEC XBRL company facts (CoreWeave Q2 2026, Equinix Q3 2026, Digital Realty Q1 2026); CoreWeave Q2 2026 earnings release (D&A, Adjusted EBITDA); '
                 'fleet.csv; 08 §3 (borrowed share, interest rate). Wear-out: CoreWeave\'s 6-year accounting life.')
fig.tight_layout(rect=(0, 0.03, 1, 1))
fig.savefig(CHARTS / '10_margins.png', dpi=200, bbox_inches='tight')
plt.show()
```

**What this shows** (`charts/10_margins.png` and the tables above):
- **CoreWeave is the reader's worst case, and then some.** It borrows at ~9.5% and its Q2 operating margin was −2%. Before depreciation the margin is 59%, which is where the "great business" impression comes from. Depreciation takes 54¢ of every revenue dollar, because the GPUs wear out in ~6 years. Then interest takes another 25¢.
- **Per $1 of equipment:** the GPUs earn ~18¢ of cash a year, ~17¢ of which is the equipment wearing out. That leaves ~1.5¢ against ~6.4¢ of interest. The gap is paid by new borrowing and new shares, as 06 found.
- **The data-center landlords are a different business.** Equinix runs a ~23% operating margin and Digital Realty ~16%, with depreciation of ~21–31% of revenue, because buildings last 20–40 years, not 6. They borrow at ~3%, and interest is ~6–7% of revenue. Even so, Digital Realty's operating profit on its assets (~2%) is below its cost of debt. Landlords make their money on cheap, long debt and buildings that hold their value.
- **The reader's long-run view is the bondholder's problem.** If GPU rental margins drift to Equinix's ~23%, CoreWeave's equipment would earn ~7% a year (23% × 31¢ of revenue per $1 of equipment) against its 9.5% borrowing cost. A GPU business can only borrow like a landlord if its assets last like a building. GPUs don't.
- **The −2% assumes the GPUs are worthless after 6 years.** §5 asks what happens if they aren't.
- **What would make the margin work:** the spot prices in §3. At today's executed B200 price, one GPU's cash per $1 is several times CoreWeave's average. CoreWeave's bonds are a bet that it can reprice its contracts toward those levels when they roll off in 2028–30, and that the shortage lasts that long.

## 5 · What if the GPUs don't burn out? Can a second life rescue CoreWeave?

**Why ask.** §4's margin is after depreciation, and depreciation is an assumption: CoreWeave writes its GPUs down to zero over 6 years. Most of the −2% is that write-down (54¢ of every revenue dollar). But GPUs don't stop working at 6. The A100 is about six and a half years old and still rents for ~$1.05–1.80 an hour (§1). If CoreWeave's chips keep earning after their contracts end, the "wear-out" in §4 is partly an accounting cost. The question is whether that second life is big enough to change the answer for bondholders.

**Method.** We follow $1 of CoreWeave equipment through its life, in cents per $1 of what it cost, per year:
- **Years 1–5, under contract:** the company average from §3: ~31¢ of revenue and ~18¢ of cash a year (Adjusted EBITDA ÷ equipment cost). Contracts average ~5 years.
- **Years 6+, second life:** nobody knows what a 2031 Blackwell will rent for, so we use the closest thing we can observe: **what a ~6-year-old chip earns today per dollar it cost new.** That's the A100: ~$27k all-in per GPU in 2020 (DGX A100 launch price ÷ 8 + networking, §2), renting now for:
  - **Glut:** $0.75, the A100's spot low in May 2026;
  - **Base:** $1.06, today's executed price (Ornn, 30-day median);
  - **High:** $1.79, today's on-demand list price;
  - **Burnout:** $0, the accounting view.

  Second-life cash an hour = rent × utilization × (1 − overhead) − running cost. Running cost is §3's power and space (~$0.28/hr for an A100). Overhead is CoreWeave's own G&A plus sales and marketing, ~9% of revenue in Q2 (SEC XBRL). Utilization: 90%, and 60% as a check.
- **Annualized and averaged,** for total lives of 5 years (dies when the contract ends), 6 (the accounting life), 8 and 10:
  - **annual return** (internal rate of return): the yearly rate at which the cash pays back the $1, comparable to the 9.5% CoreWeave pays to borrow;
  - **lifetime average operating margin:** (all the cash over the life − the $1 it cost) ÷ all the revenue, which is depreciation spread over the life the chip actually has.
- **The lender's line:** per $1 of equipment, CoreWeave borrowed ~68¢ at ~9.5% (§4), so it owes 68¢ back plus ~6.4¢ of interest every year the debt is outstanding. We ask when, if ever, cumulative cash catches that line.

**The cross-check first.** Before borrowing the A100's numbers, we check how rent per original dollar falls with age across all four generations at today's prices.

```python
# rough first-volume-shipment dates (Nvidia launch and availability announcements; ±6 months is fine for an age axis)
SHIPPED = pd.to_datetime(pd.Series({'A100': '2020-06-01', 'H100': '2023-03-01', 'H200': '2024-06-01', 'B200': '2025-01-01'}))
TODAY = ORNN['H100'].index[-1]
AGE = (TODAY - SHIPPED).dt.days / 365.25
g = facts('CRWV')
OVH = (g(['GeneralAndAdministrativeExpense'], '2026-06-30', '2026-04-01') + g(['SellingAndMarketingExpense'], '2026-06-30', '2026-04-01')) / v('crwv_rev_q2')

def cash_per_dollar(rate, util, gpu):
    # one year's cash from one GPU, per $1 of what its system cost new
    return max(rate * util * (1 - OVH) - RUN[gpu], 0) * HOURS / allin[gpu]

# rent per original dollar by age, at today's executed prices, 90% utilized
by_age = pd.DataFrame({'age (years)': AGE, 'rent per hour': NOW,
                       'revenue per $1 a year': NOW * 0.9 * HOURS / allin,
                       'cash per $1 a year': [cash_per_dollar(NOW[k], 0.9, k) for k in GENS]}).reindex(GENS)
slope = np.polyfit(by_age['age (years)'], np.log(by_age['revenue per $1 a year']), 1)[0]   # ponytail: 4-point cross-section, not a time series

A100_LOW = gd('A100', 'SPOT').min()
SECOND = {'Burnout: earns nothing after the contract': 0.0, f'Glut: A100 at its 2026 low (${A100_LOW:.2f})': A100_LOW,
          f'Base: A100 executed today (${NOW["A100"]:.2f})': NOW['A100'],
          f'High: A100 on-demand list (${rent_now.loc["A100", "on-demand list (GetDeploying)"]:.2f})': rent_now.loc['A100', 'on-demand list (GetDeploying)']}
C1, R1, YRS1 = CASH_PER_DOLLAR, REV_PER_DOLLAR, int(v('crwv_contract_years'))

def life(rate, util, total):
    # revenue and cash per $1, year by year: contract years at CoreWeave's average, then second life at an A100-like rent
    rev = [R1] * YRS1 + [rate * util * HOURS / allin['A100']] * (total - YRS1)
    cash = [C1] * YRS1 + [cash_per_dollar(rate, util, 'A100')] * (total - YRS1)
    return np.array(rev), np.array(cash)

def irr(cf):
    lo, hi = -0.99, 1.0
    for _ in range(100):
        r = (lo + hi) / 2
        lo, hi = (r, hi) if sum(c / (1 + r) ** t for t, c in enumerate(cf)) > 0 else (lo, r)
    return r
assert abs(irr([-1, 1.1]) - 0.1) < 1e-6 and abs(irr([-1, 0, 1.21]) - 0.1) < 1e-6

LIVES = [5, 6, 8, 10]
rows = {}
for name, rate in SECOND.items():
    for util in [0.9, 0.6]:
        r = {'second-life cash per $1 a year': cash_per_dollar(rate, util, 'A100')}
        for T in LIVES:
            rev, cash = life(rate, util, T)
            r[f'return a year, {T}-yr life'] = irr([-1, *cash])
            r[f'avg op. margin, {T}-yr life'] = (cash.sum() - 1) / rev.sum()
        rows[(name, f'{util:.0%}')] = r
SL = pd.DataFrame(rows).T
SL.index.names = ['second-life rent', 'utilization']

# what Q2's operating margin would have been with the same cash but a longer accounting life
DA_SHARE = M.loc['CoreWeave (Q2 2026)', 'depreciation']
restated = pd.Series({f'{n}-year life': M.loc['CoreWeave (Q2 2026)', 'operating margin'] + DA_SHARE * (1 - 6 / n) for n in [6, 8, 10]})

base = list(SECOND)[2]
OUT.update({'a100_rev_per_dollar_now': by_age.loc['A100', 'revenue per $1 a year'],
            'rev_per_dollar_decline_per_year_of_age': np.exp(slope) - 1,
            'second_life_cash_base_90': SL.loc[(base, '90%'), 'second-life cash per $1 a year'],
            'irr_burnout_5yr': SL.loc[(list(SECOND)[0], '90%'), 'return a year, 5-yr life'],
            'irr_burnout_6yr': SL.loc[(list(SECOND)[0], '90%'), 'return a year, 6-yr life'],
            'irr_base_8yr_90': SL.loc[(base, '90%'), 'return a year, 8-yr life'],
            'irr_base_8yr_60': SL.loc[(base, '60%'), 'return a year, 8-yr life'],
            'irr_glut_8yr_90': SL.loc[(list(SECOND)[1], '90%'), 'return a year, 8-yr life'],
            'irr_high_8yr_90': SL.loc[(list(SECOND)[3], '90%'), 'return a year, 8-yr life'],
            'avg_margin_base_8yr_90': SL.loc[(base, '90%'), 'avg op. margin, 8-yr life'],
            'crwv_q2_margin_at_8yr_life': restated['8-year life'], 'crwv_q2_margin_at_10yr_life': restated['10-year life']})
print(f'Overhead (G&A + sales & marketing): {OVH:.1%} of revenue. Revenue per $1 falls ~{1 - np.exp(slope):.0%} per year of chip age (cross-section).')
print(f'CoreWeave under contract: revenue {R1:.3f}, cash {C1:.3f} per $1 a year.')
print('Q2 operating margin restated to a longer accounting life (same cash):', restated.map('{:+.1%}'.format).to_dict())
display(by_age.style.format({'age (years)': '{:.1f}', 'rent per hour': '${:.2f}', 'revenue per $1 a year': '{:.2f}', 'cash per $1 a year': '{:.2f}'}),
        SL.style.format('{:.1%}').format('{:.3f}', subset=['second-life cash per $1 a year']))
```

**What this shows** (`charts/10_second_life.png` and the tables above):
- **Old GPUs earn as much as CoreWeave's new ones.** A ~6½-year-old A100 rents today for ~30¢ a year per dollar its system cost new, and keeps ~19¢ of that as cash. CoreWeave's new equipment earns 31¢ and 18¢ under contract (left panel). The point isn't that old chips are great. It's that CoreWeave's contract prices are low: a B200 renting by the hour earns ~$1.27 a year per $1, four times CoreWeave's contract rate. Across generations, rent per original dollar falls ~27% for each year of age (a four-point snapshot, not a history).
- **Depreciation is most of the "loss," so the useful life is what matters.** With the same Q2 cash, an 8-year accounting life would give CoreWeave a ~+12% operating margin, and a 10-year life ~+20%, close to Equinix's 23% (§4). The accounting change itself creates no cash. What matters is whether the rent in years 6–10 actually arrives.
- **If the chips die when the contracts end, the equipment loses money:** about −3% a year, with a −6% average lifetime margin. Five years of 18¢ is 91¢ back on each $1.
- **With a second life at today's A100 rents, the picture changes** (right panel; returns a year over an 8-year life):

  | Second life | 90% utilized | 60% utilized |
  |---|---|---|
  | Glut ($0.75) | ~5% | ~1% |
  | Base ($1.06) | ~9.4% (lifetime average margin ~19%) | ~5% |
  | High ($1.79) | ~16% | ~11% |

  A 10-year life at the base rent gives ~13%. (Revised in §7: these lives count from purchase. Azure and Google retired past chips 7–9.5 years after *launch*, and CoreWeave installs 0.5–1.5 years after launch. So an 8-year life is the top of the observed range, and 10 years is longer than any hyperscaler has run a generation.)
- **Against the 9.5% CoreWeave pays to borrow,** the base case only breaks even. The equipment earns about what the debt costs, so the shareholders' 32¢ per $1 earns roughly nothing extra. It takes high second-life rents, or near-full utilization for 10 years, to clear the cost of debt comfortably.
- **For lenders, a second life is the difference between a hole and repayment.** Per $1 of equipment, cumulative cash covers the 68¢ borrowed plus interest:
  - in year 6 at today's A100 executed price;
  - in year 7 at the 2026 low;
  - never, if the chips die with their contracts.

  GPUs bought in 2025–26 reach year 6 in 2030–32. That's when CoreWeave's unsecured notes mature (2030, 2031, 2032). So second-life rent mostly matters to those bondholders, and to whoever refinances them.
- **What it rests on:** that a 2031 Blackwell earns, per dollar it cost, what a 2026 A100 earns today. Today's A100 price comes from a year of shortage (§1). Its spot price was $0.75 six months ago and is $1.21 now. Over 2023–25, H100 rents fell ~80%.

**So:** yes, a second life can rescue CoreWeave's margin on paper and gets lenders repaid, if old-chip rents hold near today's levels. But it only brings the business to roughly its cost of debt; it doesn't make it a good one. The bigger lever is repricing: a 3.7-year-old H100 at today's executed price earns ~64¢ per $1, twice CoreWeave's contract rate. Contracts that roll off in a shortage would reprice up. Contracts that roll off in a glut would not.

## 6 · Why are CoreWeave's contracts priced so low, and can it reprice them?

**Why ask.** §3–5 keep landing on one number: CoreWeave earns ~31¢ a year per $1 of equipment, while a GPU rented on today's market earns 64¢–$1.27 (§5). Even a 6-year-old A100 earns as much per original dollar as CoreWeave's new fleet. So: were the contracts signed when prices were low? Did CoreWeave give discounts to win them? Could it earn more renting by the hour, even at poor utilization? And will the big contracts (Meta's especially) renew at higher prices on the same equipment, and run out to 10 years? That last one is the stockholders' case.

**Method.** One yardstick, **revenue per megawatt per year**, because CoreWeave reports power, not GPU counts:
- **CoreWeave's actual revenue:** quarterly revenue × 4 ÷ average active power, Q3 2025 to Q2 2026 (SEC XBRL revenue; active power from earnings releases and calls).
- **Market prices** converted to the same unit: $ per GPU-hour × 8,760 hours ÷ the GPU's power draw (kW, §3). We use executed prices (Ornn) and GetDeploying's **36-month reserved** list prices, the closest public proxy for a multi-year contract. Its 36-month figures come from only 2–7 providers, so they're noisy.
- **CoreWeave's new contracts:** its Sep 17 2026 release says short contracts (~3–6 months) signed in Q3 were priced at **~$40M per MW** of annualized revenue.
- **Timing:** each named contract's signing month (`data/neocloud_customers.csv`) against the market price that month.
- **What management has said** about pricing, renewals and old GPUs (Q4 2025 and Q2 2026 calls, the Sep 17 release; quoted in `data/neocloud_inputs.csv`).
- **The stockholders' view:** the same $1 of equipment as §5, but from the 32¢ shareholders put in. The ~68¢ loan is paid off from contract cash over the 5 years. CoreWeave's CFO said on the Q2 2026 call that "when an initial contract ends, the cluster no longer has any leverage." After that, everything the GPU earns goes to the company.

```python
# --- timing: contracts against the market price when signed ---
CUST = pd.read_csv('data/neocloud_customers.csv').query("provider == 'CoreWeave' and contract_bn.notna()")
SIGNED = pd.concat([pd.DataFrame({'who': CUST.customer, 'bn': CUST.contract_bn, 'date': pd.to_datetime(CUST.signed)}),
                    pd.DataFrame([{'who': 'Unnamed (early Q3)', 'bn': v('crwv_q3_adds'), 'date': pd.Timestamp('2026-07-15')}])]).sort_values('date')

def reserved(gpu, months):
    x = GD[(GD.gpu_slug == f'nvidia-{gpu.lower()}') & (GD.billing_type == 'RESERVATION') & (GD.reservation_months == months)]
    return x.set_index('date').provider_median_price.sort_index().resample('MS').median()
R36 = {g: reserved(g, 36) for g in ['H100', 'B200']}
h100_mkt = pd.concat([mkt, spot.resample('MS').mean().loc['2025-11':]]).resample('MS').mean().interpolate()   # §3's path
SIGNED['H100 market at signing'] = SIGNED.date.map(lambda d: h100_mkt.asof(d))
SIGNED['B200 36-mo reserved at signing'] = SIGNED.date.map(lambda d: R36['B200'].asof(d) if d >= R36['B200'].index[0] else np.nan)

# --- revenue per MW a year: CoreWeave's actual vs. the market ---
g = facts('CRWV')
REVQ = pd.Series({'Q3 2025': g(REV, '2025-09-30', '2025-07-01'),
                  'Q4 2025': g(REV, '2025-12-31', '2025-01-01') - g(REV, '2025-09-30', '2025-01-01'),
                  'Q1 2026': g(REV, '2026-03-31', '2026-01-01'), 'Q2 2026': v('crwv_rev_q2')})
GW = [v('crwv_gw_q2_2025'), v('crwv_gw_q3_2025'), v('crwv_gw_q4_2025'), v('crwv_gw_start_q2'), v('crwv_gw_end_q2')]
AVG_GW = pd.Series([(a + b) / 2 for a, b in zip(GW[:-1], GW[1:])], REVQ.index)
REALIZED = REVQ * 4 / AVG_GW            # $bn per GW = $M per MW a year
REAL_AVG = REALIZED.mean()

KW = pd.Series({k: mid(k, 'kw_per_gpu') for k in GENS})
per_mw = lambda rate, gpu: rate * HOURS / KW[gpu] / 1e3      # $M per MW a year at 100% utilization
r36_signing = R36['B200'].asof(pd.Timestamp('2026-03-01'))
YARD = pd.Series({
    f'CoreWeave actual, avg of last 4 quarters': REAL_AVG,
    f'B200 36-month reserved, Mar 2026 (${r36_signing:.2f}/hr)': per_mw(r36_signing, 'B200'),
    f'A100 executed today (${NOW["A100"]:.2f}/hr)': per_mw(NOW['A100'], 'A100'),
    f'H100 executed today (${NOW["H100"]:.2f}/hr)': per_mw(NOW['H100'], 'H100'),
    f'B200 36-month reserved, now (${R36["B200"].iloc[-1]:.2f}/hr)': per_mw(R36['B200'].iloc[-1], 'B200'),
    'CoreWeave new short contracts, Q3 2026': v('crwv_short_contract_usd_mn_per_mw'),
    f'B200 executed today (${NOW["B200"]:.2f}/hr)': per_mw(NOW['B200'], 'B200')})
BREAKEVEN_UTIL = (REAL_AVG / YARD.iloc[2:]).rename('utilization that matches CoreWeave\'s actual revenue')

# --- stockholders: 32¢ in, loan paid off from contract cash over 5 years, then the second life is theirs ---
PMT = LOAN * RATE / (1 - (1 + RATE) ** -YRS1)       # level loan payment a year, per $1 of equipment
def equity_irr(rate, util, total):
    _, cash = life(rate, util, total)
    return irr([-(1 - LOAN), *(cash - np.r_[[PMT] * YRS1, [0] * (total - YRS1)])])
EQ = pd.DataFrame({f'{T}-year life': {n: equity_irr(r, 0.9, T) for n, r in SECOND.items()} for T in [6, 8, 10]})
left_in_contract = C1 - PMT

OUT.update({'crwv_rev_per_mw_actual': REAL_AVG, 'crwv_rev_per_mw_new_short': v('crwv_short_contract_usd_mn_per_mw'),
            'b200_36mo_reserved_mar2026': r36_signing, 'b200_36mo_reserved_now': R36['B200'].iloc[-1],
            'breakeven_util_b200_executed': BREAKEVEN_UTIL.iloc[-1], 'breakeven_util_h100_executed': BREAKEVEN_UTIL.iloc[1],
            'equity_irr_burnout': EQ.iloc[0, 1], 'equity_irr_base_8yr': EQ.iloc[2, 1], 'equity_irr_base_10yr': EQ.iloc[2, 2],
            'equity_irr_glut_10yr': EQ.iloc[1, 2], 'equity_irr_high_10yr': EQ.iloc[3, 2]})
print(f'CoreWeave revenue per MW a year ($M): {REALIZED.round(1).to_dict()}; average {REAL_AVG:.1f}')
print(f'Loan payment {PMT:.3f} per $1 a year vs. contract cash {C1:.3f}: {left_in_contract * 100:.1f}¢ a year left for shareholders during the contract.')
display(SIGNED.style.format({'bn': '${:.1f}B', 'date': '{:%b %Y}', 'H100 market at signing': '${:.2f}', 'B200 36-mo reserved at signing': '${:.2f}'}, na_rep='—').hide(axis='index'),
        YARD.round(1).to_frame('$M per MW a year (100% utilized)'), BREAKEVEN_UTIL.map('{:.0%}'.format).to_frame(),
        EQ.style.format('{:.1%}').set_caption('Shareholders\' return a year on their 32¢, second life at 90% utilization'))
```

**What this shows** (`charts/10_contracts.png` and the tables above):

**1. Were the contracts signed in a glut? Mostly, yes.**
- ~$64B of the ~$89B of contracts with known amounts were signed between Mar 2025 and Apr 2026. Over that period, the H100's market price fell from ~$2.30 to ~$1.50 an hour.
- Meta's $21B contract was signed in Mar 2026, the month the market bottomed. A 36-month B200 reservation was then listed at ~$2.79 an hour. It's ~$7.16 now, 2.6× higher, and B200 executed prices are up ~70% since August.
- Only the >$25B signed in early Q3 2026 came after prices turned.
- An investor (Gavin Baker, Oct 2026) put it this way: CoreWeave "locked in a vast amount of compute" in contracts "that are now way, way under market."

**2. Did they discount to win them? Not in any way they disclose. They priced to cost, not to the market.**
- CoreWeave earns ~$9M a year per MW of active power. The quarterly figure slipped from ~$10M to ~$8M as new halls came online before customers started paying.
- That's about what a 36-month B200 reservation fetched when Meta signed (~$14M per MW at 100% utilization, less a volume discount for the biggest buyer), and about what today's A100 earns.
- So the contracts were roughly market-priced when signed. The market has since moved.
- Management describes pricing as a target margin on the equipment's cost (Q4 2025 call): "as the infrastructure changes in price, we are altering our pricing… to target the type of margins" it wants. Mature contracts run ~70% EBITDA margins.
- The concessions are in the terms, not the headline price. OpenAI got ~$350M of CoreWeave stock alongside its first $11.9B contract, ~3% of its value. Management also calls prepayment "a lever… to change the economics around a contract."
- This is a lessor's model: sell 5 years of fixed-price capacity, borrow against it, and pay the loan off within the term. The customer keeps the upside if prices rise.

**3. Could CoreWeave earn more renting by the hour, even at poor utilization? At today's prices, by a lot.**
- Today's executed B200 price matches CoreWeave's revenue per MW at ~20% utilization. The H100 matches it at ~42%, and CoreWeave's own new 3–6 month contracts ($40M per MW) at ~23%.
- CoreWeave is moving that way: a ~25% price increase across products in July, more short and 2–3-year contracts, and new contracts at 5–10 points higher margins.
- But the shortcut doesn't scale, for three reasons:
  - **Market size.** The hourly market is a small market at the margin. CoreWeave's ~1.5 GW is on the order of a million GPUs, and dumping it there would move the price.
  - **Timing.** The comparison uses today's shortage prices. At Mar 2026 prices, the gap was ~1.5×, not 4–5×.
  - **Financing.** The contracts are what made the borrowing possible. Lenders lent ~68¢ per $1 against take-or-pay cash from Meta and Microsoft, not against hourly rentals. **CoreWeave traded the upside for the financing.** That's why it's paid so little, and why it could build this fast.

**4. Will Meta renew at a higher price on the same equipment? Not before 2031–32, and then it depends on that year's market.**
- Meta's contracts run to Dec 2031 ($14.2B) and Dec 2032 ($21B). That's ~6 years, about the GB300's whole accounting life, at fixed prices. July's increase doesn't touch them.
- A renewal would be for 6-year-old chips. The evidence on old chips is good:
  - H100s coming off an expired contract were rebooked at ~95% of the original price (the CEO to CNBC). Those originals were signed at 2023–24 shortage prices.
  - An A100 contract (a 2020 chip) runs into 2029 "at full freight": a 9-year working life.
- But the CFO says renewals are still "a very limited part of our fleet." And Meta is building its own campuses (07), so it may simply not renew.

**5. The stockholders' case: a 10-year life is where the money is.**
- During the contract, loan payments (~17.6¢ a year per $1) absorb almost all the cash (~18.2¢). Shareholders get ~0.6¢ a year on their 32¢.
- After the contract, the equipment carries no debt and every dollar is theirs. Their return a year (second life at 90% utilization):

  | After the contract | 8-year life | 10-year life |
  |---|---|---|
  | Chips die when the contract ends | −48% | −48% |
  | Glut ($0.75) | ~1% | ~7.5% |
  | Base: A100's executed price today ($1.06) | ~9% | **~15%** |
  | High ($1.79) | ~21% | ~26% |

- So **the stock is a bet on 2030–36 old-chip rents and lifespans,** and the A100-to-2029 contract is the best evidence for it.
- The bondholders' version is the same bet with a capped payoff. The unsecured notes (2030–32) are repaid or refinanced in exactly the years when that second-life cash starts arriving.

**Caveats.**

- The 36-month reserved prices come from only 2–7 providers.
- The stockholders' view treats all debt as paid off from contract cash, as CoreWeave says of its asset-level loans. Its unsecured bullet notes are refinanced instead.

## 7 · How long will CoreWeave's GPUs last, and what is the time after the contract worth?

**Why ask.** §5–6 show the stock's case rests on one number: how long a GPU keeps earning after its first contract ends. CoreWeave writes its equipment off over 6 years, and its contracts run ~5. If the chips work for 10 years, the last 4–5 are nearly all profit: the loans are repaid (§6), there's no depreciation left, and CoreWeave owns the chips outright. That could be most of what the stock is worth. Two questions: what does CoreWeave own (how many MW of each chip generation, installed when), and how long has each past generation lasted?

**Method.**
- **Fleet by generation.** CoreWeave doesn't disclose its fleet by chip. It does report **active power** each quarter (70 MW at Dec 2023, 360 MW at Dec 2024, then quarterly to 1.5 GW at Jun 2026; `data/neocloud_inputs.csv`). Each quarter's added MW is assigned to the chips CoreWeave was installing then. Those dates are public because CoreWeave announced being first with each one: H100 in 2023 (incl. Inflection's 22,000-GPU cluster), H200 from mid-2024, GB200 from early 2025, GB300 from July 2025, Rubin in H2 2026. The split within a quarter is our assumption, shown in the code. We assume nothing has been retired, because management says its A100s and H100s are "sold out" (Q1 2026 call).
- **How long past generations lasted.** Two kinds of evidence (`data/gpu_lifespans.csv`):
  - **When the big clouds retired them**, measured from the chip's launch: Azure retired its K80, P40 and P100 machines in Aug 2023 and its V100 machines in Sep 2025. Google ended K80 support in May 2024. This is the stricter test: a hyperscaler retires a chip when its power and floor space can earn more with something newer, not when the chip stops working.
  - **Whether anyone still rents them.** GetDeploying's Nov 2026 listings show K80s (12 years old), P100s, V100s and T4s still offered. This is a ceiling: some providers still list them, but the volume is small.
- **Each CoreWeave slice's timeline.** Installed in its quarter. Contract ends ~5 years later (the weighted average, 10-K). Accounting life ends at 6 years. Working life ends when the chip reaches the hyperscaler retirement age, counted from its *launch*: a cloud retires a whole generation at once, so a chip installed late gets less of a life.
- **What the second life is worth.** CoreWeave's $33.8B of technology equipment (Jun 2026, `fleet.csv`) is split across the slices by MW. After its contract, each $1 earns §5's second-life cash: an A100-like rent, 90% utilized, after power, space and overhead. Two cases: §5's base (A100 executed today) and glut (the A100's 2026 low). That cash is discounted to today at 12%, a rough equity hurdle (the secured debt costs ~9.5%), and compared with the stock's market value. This counts only the equipment installed by Jun 2026; the 2.7 GW contracted but not yet built would add its own second life, years later.

```python
from common import load_equity, SURFACE

# --- 1. CoreWeave's active power by quarter (MW), split by chip generation ---
POWER = pd.Series({'2023-12-31': v('crwv_gw_q4_2023'), '2024-12-31': v('crwv_gw_q4_2024'), '2025-03-31': v('crwv_gw_q1_2025'),
                   '2025-06-30': v('crwv_gw_q2_2025'), '2025-09-30': v('crwv_gw_q3_2025'), '2025-12-31': v('crwv_gw_q4_2025'),
                   '2026-03-31': v('crwv_gw_start_q2'), '2026-06-30': v('crwv_gw_end_q2'), '2026-12-31': v('crwv_gw_ye2026_guide')}) * 1000
POWER.index = pd.to_datetime(POWER.index)
FLEET_GENS = ['Ampere', 'H100', 'H200', 'GB200', 'GB300', 'Rubin']
LAUNCH = pd.to_datetime(pd.Series({'Ampere': SHIPPED['A100'], 'H100': SHIPPED['H100'], 'H200': SHIPPED['H200'],
                                   'GB200': SHIPPED['B200'], 'GB300': '2025-07-01', 'Rubin': '2026-09-01'}))
# Dec 2023's 70 MW: Inflection's 22,000 H100s are Hopper; the rest Ampere-era (A100, A40, RTX).
# Active power is 'critical IT load' (CFO, Q3 2026 call), so no cooling multiplier.
hop23 = v('crwv_hopper_gpus_2023') * v('h100_kw') / 1000
# ponytail: share of each period's added MW by chip, set from CoreWeave's "first to deploy" dates; ±1 quarter is the honest precision
MIX = {'2024-12-31': {'H100': .7, 'H200': .3},        # H200 from mid/late 2024
       '2025-03-31': {'H200': .5, 'GB200': .5},       # GB200 NVL72 generally available early 2025
       '2025-06-30': {'GB200': 1},
       '2025-09-30': {'GB200': .8, 'GB300': .2},      # first GB300 NVL72 Jul 3 2025
       '2025-12-31': {'GB200': .5, 'GB300': .5},
       '2026-03-31': {'GB200': .2, 'GB300': .8},
       '2026-06-30': {'GB300': 1},
       '2026-12-31': {'GB300': .5, 'Rubin': .5}}       # guidance; Rubin "second half of 2026" (10-Q)
rows = [{'gen': 'Ampere', 'installed': pd.Timestamp('2022-07-01'), 'mw': POWER.iloc[0] - hop23, 'end': POWER.index[0]},
        {'gen': 'H100', 'installed': pd.Timestamp('2023-10-01'), 'mw': hop23, 'end': POWER.index[0]}]
for prev, end in zip(POWER.index[:-1], POWER.index[1:]):
    assert abs(sum(MIX[str(end.date())].values()) - 1) < 1e-9
    for gen, w in MIX[str(end.date())].items():
        rows.append({'gen': gen, 'installed': prev + (end - prev) / 2, 'mw': (POWER[end] - POWER[prev]) * w, 'end': end})
SLICES = pd.DataFrame(rows)
SLICES['guided'] = SLICES.end > pd.Timestamp('2026-06-30')
STACK = pd.DataFrame({d: SLICES[SLICES.end <= d].groupby('gen').mw.sum() for d in POWER.index}).T.reindex(columns=FLEET_GENS).fillna(0)
assert np.allclose(STACK.sum(axis=1), POWER)

# --- 2. How long past generations lasted (years from launch) ---
LS = pd.read_csv('data/gpu_lifespans.csv', parse_dates=['launched', 'date'])
LS['age'] = (LS.date - LS.launched).dt.days / 365.25
RET = LS[LS.event == 'hyperscaler_retired']
L_LOW, L_BASE, L_HIGH = RET.age.min(), RET.age.median(), RET.age.max()
launched = LS.drop_duplicates('gpu').set_index('gpu').launched
STILL = pd.DataFrame({k: {'age now': (TODAY - launched[k]).days / 365.25,
                          'providers listing (on-demand)': GD[(GD.gpu_slug == f'nvidia-{k.lower()}') & (GD.billing_type == 'ON_DEMAND') & (GD.date == GD.date.max())].provider_count.sum(),
                          'on-demand price': gd(k, 'ON_DEMAND').iloc[-1]} for k in ['K80', 'P100', 'V100', 'T4', 'A100']}).T

# --- 3. Each slice's timeline ---
yrs = lambda d: pd.to_timedelta(d * 365.25, unit='D')
BOOK = pd.read_csv('data/fleet.csv').query("issuer == 'CoreWeave' and asset_class.str.startswith('Technology')").iloc[0]
SLICES['launch'] = SLICES.gen.map(LAUNCH)
SLICES['contract_end'] = SLICES.installed + yrs(YRS1)
SLICES['book_end'] = SLICES.installed + yrs(BOOK.life_high_yrs)
SLICES['life_end'] = SLICES.launch + yrs(L_BASE)
SLICES['after contract (yrs)'] = ((SLICES.life_end - SLICES.contract_end).dt.days / 365.25).clip(lower=0)
built = SLICES[~SLICES.guided]
wavg = lambda s, col: s[col].min() + pd.to_timedelta(np.average((s[col] - s[col].min()).dt.days, weights=s.mw), unit='D')
BYGEN = pd.DataFrame({gen: {'MW (Jun 2026)': s.mw.sum(), 'share': s.mw.sum() / built.mw.sum(),
                            'installed (avg)': wavg(s, 'installed'),
                            'contract ends': wavg(s, 'contract_end'),
                            'written off': wavg(s, 'book_end'),
                            f'retired at {L_BASE:.1f} yrs': s.life_end.iloc[0],
                            **{f'yrs after contract, {L:.1f}-yr life': np.average(((s.launch + yrs(L) - s.contract_end).dt.days / 365.25).clip(lower=0), weights=s.mw)
                               for L in [L_LOW, L_BASE, L_HIGH]}}
                      for gen, s in built.groupby('gen', sort=False)}).T

# --- 4. What the second life is worth today ---
DISC = v('discount_rate_equity')
COST_PER_MW = BOOK.gross_bn / built.mw.sum()                # $bn of technology equipment per MW
CASE = {'Base': SECOND[base], 'Glut': SECOND[list(SECOND)[1]]}
def second_life_pv(L, rate, util=0.9, disc=DISC):
    c = cash_per_dollar(rate, util, 'A100')
    t0 = ((built.contract_end - TODAY).dt.days / 365.25).clip(lower=0)
    t1 = ((built.launch + yrs(L) - TODAY).dt.days / 365.25).clip(lower=t0)
    k = np.log(1 + disc)                                      # cash spread evenly from contract end to retirement
    return (built.mw * COST_PER_MW * c * ((1 + disc) ** -t0 - (1 + disc) ** -t1) / k).sum()
assert abs(second_life_pv(L_BASE, CASE['Base'], disc=1e-9) -
           (built.mw * COST_PER_MW * cash_per_dollar(CASE['Base'], 0.9, 'A100') * built['after contract (yrs)']).sum()) < 1e-3

ISS = pd.read_csv('issuers.csv').set_index('ticker').loc['CRWV']
px = load_equity('CRWV').close
MCAP = px.iloc[-1] * ISS.shares_mn / 1e3
LIFE_GRID = np.arange(6, 12.01, 0.25)
CURVE = pd.DataFrame({n: [second_life_pv(L, r) for L in LIFE_GRID] for n, r in CASE.items()}, index=LIFE_GRID)
PV = pd.DataFrame({n: {f'{L:.1f}-yr life': second_life_pv(L, r) for L in [L_LOW, L_BASE, L_HIGH, 10.0]} for n, r in CASE.items()})
undisc = second_life_pv(L_BASE, CASE['Base'], disc=1e-9)

OUT.update({'life_hyperscaler_low': L_LOW, 'life_hyperscaler_median': L_BASE, 'life_hyperscaler_high': L_HIGH,
            'crwv_fleet_share_blackwell_jun2026': BYGEN.loc[['GB200', 'GB300'], 'share'].sum(),
            'crwv_fleet_share_hopper_jun2026': BYGEN.loc[['H100', 'H200'], 'share'].sum(),
            'crwv_yrs_after_contract_median_life': np.average(built['after contract (yrs)'], weights=built.mw),
            'crwv_second_life_pv_base_bn': PV.loc[f'{L_BASE:.1f}-yr life', 'Base'], 'crwv_second_life_pv_glut_bn': PV.loc[f'{L_BASE:.1f}-yr life', 'Glut'],
            'crwv_second_life_pv_base_10yr_bn': PV.loc['10.0-yr life', 'Base'], 'crwv_market_cap_bn': MCAP})
print(f'Hyperscaler retirement age (years from launch): low {L_LOW:.1f}, median {L_BASE:.1f}, high {L_HIGH:.1f}')
print(f'Second-life cash per $1 a year (90%): base {cash_per_dollar(CASE["Base"], 0.9, "A100"):.3f}, glut {cash_per_dollar(CASE["Glut"], 0.9, "A100"):.3f}; under contract {C1:.3f}')
aft = {L: np.average(((built.launch + yrs(L) - built.contract_end).dt.days / 365.25).clip(lower=0), weights=built.mw) for L in [L_LOW, L_BASE, L_HIGH, 11.0]}
print('Years after contract, MW-weighted, by life from launch:', {f'{k:.1f}': round(x, 2) for k, x in aft.items()})
print(f'Base value at the {RATE:.1%} cost of debt instead of {DISC:.0%}: ${second_life_pv(L_BASE, CASE["Base"], disc=RATE):.1f}B')
print(f'Technology equipment ${BOOK.gross_bn:.1f}B over {built.mw.sum():.0f} MW = ${COST_PER_MW * 1e3:.1f}M per MW')
print(f'CoreWeave market value ${MCAP:.1f}B ({ISS.shares_mn:.0f}M shares × ${px.iloc[-1]:.2f}, {px.index[-1]:%b %d %Y}). '
      f'Second life undiscounted (base, {L_BASE:.1f} yrs): ${undisc:.1f}B')
display(RET[['gpu', 'provider', 'launched', 'date', 'age']].style.format({'launched': '{:%b %Y}', 'date': '{:%b %Y}', 'age': '{:.1f}'}).hide(axis='index'),
        STILL.style.format({'age now': '{:.1f}', 'providers listing (on-demand)': '{:.0f}', 'on-demand price': '${:.2f}'}),
        BYGEN.style.format({'MW (Jun 2026)': '{:.0f}', 'share': '{:.0%}', 'installed (avg)': '{:%b %Y}', 'contract ends': '{:%b %Y}',
                            'written off': '{:%b %Y}', f'retired at {L_BASE:.1f} yrs': '{:%b %Y}'} |
                           {c: '{:.1f}' for c in BYGEN.columns if c.startswith('yrs')}),
        PV.style.format('${:.1f}B').set_caption(f'Second life of the Jun 2026 fleet, discounted to today at {DISC:.0%}'))
```

**1. What CoreWeave owns: mostly Blackwell, mostly installed in the last 18 months (left).** At Jun 30 2026, our estimate of the 1.5 GW is:

|---|---|---|
| GB300 | ~775 | 52% |

So ~¾ is Blackwell, and only ~3% is old enough to tell us anything about a second life. This split is our estimate from install dates; CoreWeave doesn't disclose it.

**2. How long chips last: the big clouds retire them 7–9.5 years after launch (middle).**
- Azure retired its P40 and P100 machines at ~7 years and its V100s at ~8.3. Azure and Google retired the K80 at 8.8–9.5 years. The median is **8.3 years**.
- That's the hyperscalers' choice, not the chip's death. V100s (9.4 years old) are still listed by 17 providers today, P100s by 4 and a K80 by 1.
- CoreWeave's A100 contract "into 2029" would put that chip at ~9 years, at the top of the range.
- Clouds retire chips when the power and floor space earn more with newer ones. Per §6, a MW of A100s earns ~$11M a year at today's prices and a MW of B200s ~$46M.

**3. The catch: the clock starts at launch, not at installation, and CoreWeave installs late (middle, CoreWeave rows).**
- CoreWeave installs each chip 0.5–1.5 years after launch, then sells a ~5-year contract.
- So at the median retirement age, the fleet gets **~2.5 years after its contracts** (MW-weighted) and ~1.5 years after it's written off.
- At the shortest observed life (7 years) that's ~1.2 years. At the longest (9.5 years) it's ~3.6.
- The 4–5 extra years of a "10-year GPU" need a life of ~11 years from launch. No hyperscaler has run a generation that long.

**4. What the extra years are worth: ~$9B today, ~33% of the stock (right).** (Revised in §8: valued per MW of power instead of per dollar of equipment, and with the ~300k GPUs management says come off contract in 2027–28, the base case is ~$16B, ~60% of the stock's value.)
- The Jun 2026 fleet's second life is worth **~$8.9B** today at §5's base rent, the A100's executed price today ($1.06), 90% utilized, discounted at 12%. It's ~$5.1B at the 2026 glut price.
- The stock's market value is ~$27B ($48.73, Nov 10 close; ~$30B at Oct 23's $54.24). With a 10-year life the base value is ~$14B, ~52% of the stock.
- Undiscounted, the base case is ~$15.7B of cash, arriving between 2027 and 2033.
- Within 6–12 year lives, the existing fleet's second life never reaches the stock's value. The rest has to come from the 2.7 GW contracted but not yet built, and from new contracts at today's higher prices (§6).

**5. Is a second life "much better than the contracts"? Per dollar of cash, no. Per dollar to shareholders, yes.**
- At today's A100-like rent, a post-contract chip makes ~18.7¢ of cash a year per $1 it cost, almost exactly what it made under contract (18.2¢, §5). At the glut price it makes ~10.7¢, less than under contract.
- The gain is what the cash no longer pays for. Under contract, ~17.6¢ of the 18.2¢ goes to the lender (§6). After the contract there's no loan payment and no depreciation, so all of it is profit to the company.
- That is the bull case for the stock, and it holds. But it buys ~2.5 years of that, not 5, unless CoreWeave runs its chips longer than any hyperscaler has.

**Caveats.**
- The fleet split is an estimate from public "first to deploy" dates (±1 quarter), and it assumes nothing has been retired.
- The retirement ages are 5 data points from 2 hyperscalers.
- Second-life cash assumes a 2031–34 chip earns, per original dollar, what an A100 earns today, at 90% utilization. That's §5's assumption, and a shortage-year price.
- The 12% discount rate is a judgment call; at the 9.5% cost of debt, the $9B rises to ~$10B.
- The value is pre-tax and covers only equipment already installed.

## 8 · After Q3: is the "money printer" primed, or is CoreWeave borrowing to the edge again?

**Why ask.** §5–7 say the stock's upside is the second life: once a contract ends, the loan behind it is repaid and every dollar the chip earns is profit. CoreWeave reported Q3 2026 on Nov 10, beating its own guidance (revenue $3.66B, adjusted operating margin 10%). On the call, management priced new contracts far above the old book and laid out its renewal plans. Three questions:
1. **Did Q3 change the economics?** Revenue per MW, margins, interest, debt.
2. **Does pricing new contracts higher, and borrowing as much against them, set up a worse outcome?** The worry: borrow heavily against shortage-priced contracts, the contracts end while prices are lower, and the debt is still there. That would be worse than today's book, where the loans are gone before the second life starts.
3. **Who gets the second life's cash?** The contract loans are repaid from the contracts. But CoreWeave, Inc. itself, the listed holding company, has also borrowed in bonds that are repaid in one lump at maturity. Those mature in the same years the second life pays out.

**Who borrows what.** The GPU loans (DDTLs) are borrowed by CoreWeave's wholly owned subsidiaries (e.g. *CoreWeave Compute Acquisition Co. II, LLC* for DDTL 1.0). Each is secured on that subsidiary's GPUs and customer contracts and repaid from the contract's cash. Cash reaches the holding company only after those loans are paid. CoreWeave, Inc. also guarantees the GPU loans ("Parent Guarantee and Pledge Agreements"), except DDTL 4.0 (Meta's contract), which is non-recourse. The unsecured notes and convertibles are CoreWeave, Inc.'s own debt. Nobody owns CoreWeave, Inc. in this sense except its shareholders, so "holding company" below means CoreWeave, Inc.

**Method.**
- **Q3 vs Q2:** the Q3 release (income statement, balance sheet, cash flow) and call, entered in `data/neocloud_inputs.csv`. The 10-Q isn't filed yet.
- **Four ways to finance 1 MW**, each from what management or the loan documents say. They differ in contract price, contract length, how much of the cost is borrowed, the loan's rate and the loan's term:
  - **Old book:** today's average revenue per MW, a 5-year contract and a 68% loan at 9.5% (§4).
  - **New long:** $22.5M per MW on a 4–5-year contract (CEO/CFO: "$20–25M"), with an investment-grade loan for >90% of the cost at ~7.25% (CFO).
  - **New short (DDTL 5.5):** $40M per MW on ~3-year contracts, but ~5-year debt at SOFR + 5.5%. The release says CoreWeave may "renew existing contracts or re-lease capacity" in between. The 70% advance rate comes from a secondary source.
  - **The worry case:** the same short contracts, borrowed to 100% of cost, the top of the CEO's "70–100%".
- **Each MW:** cost is CoreWeave's equipment per MW (§3, ~$26.5M). Fixed cash costs (power, leases, staff) are Q3 revenue minus adjusted EBITDA, per MW (~$3.5M a year). The loan is repaid in level payments over its term. After the contract, the MW rents at 90% utilization at three prices per MW:
  - **Glut:** the A100's 2026 low, ~$8M per MW.
  - **Base:** the A100's executed price today, ~$11M.
  - **High:** the H100's executed price today, ~$21M. That's a ~4-year-old chip in a shortage, about what DDTL 5.5's chips will be when their contracts end.
- **Management's renewal claim:** ~130k GPUs come off contract in 2027 and ~170k in 2028, adding ~$4B and then ~$10B of annual revenue. Converted to the price per GPU-hour that implies.
- **Who owns the second life:** the holding company's bonds that come due in one lump (`data/crwv_debt.csv`, from the Q2 10-Q debt table plus Q3's new convertible), by year. Against them, the second-life cash of the Jun 2026 fleet by year, from §7, with contract ends updated: the 300k GPUs management says roll off in 2027–28 are assigned to the oldest (pre-2025) slices.

```python
SOFR = pd.read_csv('data/fred/SOFR.csv', index_col=0, parse_dates=True).SOFR.iloc[-1] / 100
gw_q3 = (v('crwv_gw_end_q2') + v('crwv_gw_end_q3')) / 2

# --- 1. Q3 against Q2 ---
def qcol(rev, ebitda, op, intr, da, gw_avg, debt, ppe_net, gw_end):
    return {'revenue ($B)': rev, 'adj. EBITDA margin': ebitda / rev, 'operating margin (GAAP)': op / rev, 'depreciation ÷ revenue': da / rev,
            'interest ÷ revenue': intr / rev, 'revenue per MW ($M a year)': rev * 4 / gw_avg, 'debt ($B)': debt,
            'debt ÷ net equipment': debt / ppe_net, 'active power (GW, end)': gw_end}
rev2 = v('crwv_rev_q2')
QQ = pd.DataFrame({
    'Q2 2026': qcol(rev2, v('crwv_adj_ebitda_q2'), M.loc['CoreWeave (Q2 2026)', 'operating margin'] * rev2, v('crwv_interest_q2'), DA_SHARE * rev2,
                    AVG_GW['Q2 2026'], v('crwv_debt_q2'), v('crwv_ppe_net_q2'), v('crwv_gw_end_q2')),
    'Q3 2026': qcol(v('crwv_rev_q3'), v('crwv_adj_ebitda_q3'), v('crwv_opinc_q3'), v('crwv_interest_q3'), v('crwv_da_q3'),
                    gw_q3, v('crwv_debt_q3'), v('crwv_ppe_net_q3'), v('crwv_gw_end_q3'))})
EBITDA_LESS_INT = (v('crwv_adj_ebitda_q3') - v('crwv_interest_q3')) * 4
FIXED = (v('crwv_rev_q3') - v('crwv_adj_ebitda_q3')) * 4 / gw_q3          # $M per MW a year of cash costs
K = cost_gw                                                                 # $M of equipment per MW (§3)

# --- 2. Four ways to finance one MW ---
lag = np.average((built.installed - built.launch).dt.days / 365.25, weights=built.mw)
LIFE_I = L_BASE - lag                                                       # §7's median life, counted from installation
SECOND_MW = {'Glut': per_mw(A100_LOW, 'A100'), 'Base': per_mw(NOW['A100'], 'A100'), 'High': per_mw(NOW['H100'], 'H100')}
S8 = {'Old book': dict(P=QQ.loc['revenue per MW ($M a year)', 'Q3 2026'], Tc=YRS1, adv=LOAN, r=RATE, Tl=YRS1),
      'New long': dict(P=v('crwv_long_contract_usd_mn_per_mw'), Tc=YRS1, adv=v('crwv_ig_ddtl_advance'), r=v('crwv_ig_ddtl_yield'), Tl=YRS1),
      'New short (DDTL 5.5)': dict(P=v('crwv_short_contract_q3_call'), Tc=v('ddtl55_contract_years'), adv=v('ddtl55_advance'),
                                   r=SOFR + v('ddtl55_spread'), Tl=v('ddtl55_maturity_years')),
      'Worry case (100% borrowed)': dict(P=v('crwv_short_contract_q3_call'), Tc=v('ddtl55_contract_years'), adv=v('crwv_ddtl_advance_high'),
                                               r=SOFR + v('ddtl55_spread'), Tl=v('ddtl55_maturity_years'))}

def mw_path(P, Tc, adv, r, Tl, P2, util2=0.9, T=LIFE_I):
    t = np.arange(1, int(np.ceil(T)) + 1)
    w = np.minimum(1, T - (t - 1))                                          # last year is partial
    cash = np.where(t <= Tc, P - FIXED, P2 * util2 - FIXED) * w
    pay = adv * K * r / (1 - (1 + r) ** -Tl)
    debt = np.where(t <= Tl, pay, 0.0)
    return t, cash, debt
def balance(adv, r, Tl, t):
    L, pay = adv * K, adv * K * r / (1 - (1 + r) ** -Tl)
    return L * (1 + r) ** t - pay * ((1 + r) ** t - 1) / r
assert abs(balance(0.7, 0.08, 5, 5)) < 1e-9

rows = {}
for name, s in S8.items():
    pay = s['adv'] * K * s['r'] / (1 - (1 + s['r']) ** -s['Tl'])
    r_ = {'price ($M/MW)': s['P'], 'contract (yrs)': s['Tc'], 'borrowed (% of cost)': s['adv'], 'loan rate': s['r'],
          'loan payment ($M/MW a yr)': pay, 'coverage during contract (×)': (s['P'] - FIXED) / pay,
          "shareholders' cash during contract ($M/MW)": (s['P'] - FIXED - pay) * s['Tc'],
          'debt left when contract ends ($M/MW)': max(balance(s['adv'], s['r'], s['Tl'], s['Tc']), 0)}
    for n2, p2 in SECOND_MW.items():
        _, cash, debt = mw_path(**s, P2=p2)
        gap = cash - debt
        r_[f'after-contract shortfall, {n2.lower()} ($M/MW)'] = float(-gap[gap < 0].sum()) + 0.0
        r_[f'lifetime cash to shareholders, {n2.lower()} ($M/MW)'] = gap.sum()
    rows[name] = r_
FIN = pd.DataFrame(rows).T
# what borrowing "the same coverage" (1.2×, the DDTL standard) against a $40M contract would mean
DSCR = v('ddtl_dscr')
s = S8['New short (DDTL 5.5)']
dscr_adv = (s['P'] - FIXED) / DSCR * (1 - (1 + s['r']) ** -s['Tc']) / s['r'] / K

# --- 3. Management's renewal claim, per GPU-hour ---
gpus = v('crwv_offcontract_gpus_2027') + v('crwv_offcontract_gpus_2028')
RENEW = pd.Series({'2027 cohort': v('crwv_renewal_rev_2027') * 1e9 / (v('crwv_offcontract_gpus_2027') * HOURS),
                   '2027 + 2028 cohorts': v('crwv_renewal_rev_2027_28') * 1e9 / (gpus * HOURS)})   # incremental $/GPU-hour
RENEW_PRICE = RENEW + implied['H100']                                       # on top of the old book's ~$1.35

# --- 4. Who owns the second life: parent bullets vs. second-life cash by year ---
DEBT = pd.read_csv('data/crwv_debt.csv', parse_dates=['maturity'])
BULLETS = DEBT[DEBT.kind == 'holding-company bullet'].groupby(DEBT.maturity.dt.year).amount_bn.sum()
sl = built.copy()
old = sl.installed < pd.Timestamp('2025-01-01')
cum = sl[old].sort_values('installed').mw.cumsum() / sl[old].mw.sum()
sl.loc[cum.index, 'contract_end'] = np.where(cum <= v('crwv_offcontract_gpus_2027') / gpus, pd.Timestamp('2027-07-01'), pd.Timestamp('2028-07-01'))
YEARS = np.arange(2027, 2036)
def by_year(slices, L, cash_per_mw):
    out = pd.Series(0.0, YEARS)
    for _, r in slices.iterrows():
        a, b = r.contract_end, r.launch + yrs(L)
        for y in YEARS:
            lo, hi = max(a, pd.Timestamp(f'{y}-01-01')), min(b, pd.Timestamp(f'{y + 1}-01-01'))
            out[y] += r.mw * cash_per_mw * max((hi - lo).days, 0) / 365.25 / 1e3          # $bn
    return out
SL_CASH = pd.DataFrame({n: by_year(sl, L_BASE, p2 * 0.9 - FIXED) for n, p2 in SECOND_MW.items()})
SL_PV = (SL_CASH.mul((1 + DISC) ** -(YEARS + 0.5 - (TODAY.year + TODAY.day_of_year / 365.25)), axis=0)).sum()
built_after_revised = np.average(((sl.launch + yrs(L_BASE) - sl.contract_end).dt.days / 365.25).clip(lower=0), weights=sl.mw)

OUT.update({'crwv_q3_rev_per_mw': QQ.loc['revenue per MW ($M a year)', 'Q3 2026'], 'crwv_q3_adj_ebitda_margin': QQ.loc['adj. EBITDA margin', 'Q3 2026'],
            'crwv_q3_interest_share': QQ.loc['interest ÷ revenue', 'Q3 2026'], 'crwv_q3_debt_to_net_ppe': QQ.loc['debt ÷ net equipment', 'Q3 2026'],
            'crwv_fixed_cash_cost_per_mw': FIXED, 'crwv_parent_bullets_bn': BULLETS.sum(),
            'crwv_second_life_cash_base_bn_revised': SL_CASH['Base'].sum(), 'crwv_second_life_cash_glut_bn_revised': SL_CASH['Glut'].sum(),
            'crwv_second_life_cash_high_bn_revised': SL_CASH['High'].sum(), 'crwv_second_life_pv_base_bn_revised': SL_PV['Base'], 'crwv_second_life_pv_glut_bn_revised': SL_PV['Glut'],
            'crwv_renewal_implied_price_hr': RENEW_PRICE.iloc[-1], 'worry_case_shortfall_glut': FIN.loc['Worry case (100% borrowed)', 'after-contract shortfall, glut ($M/MW)']})
print(f'SOFR {SOFR:.2%}. Fixed cash costs ${FIXED:.1f}M per MW a year; equipment ${K:.1f}M per MW; life from installation {LIFE_I:.1f} yrs (§7 median {L_BASE:.1f} − {lag:.1f} install lag).')
print(f'Q3 adj. EBITDA − interest, annualized: ${EBITDA_LESS_INT:.1f}B. Q3 debt raised ${v("crwv_debt_issued_q3"):.1f}B, repaid ${v("crwv_debt_repaid_q3"):.1f}B; prepayments +${v("crwv_deferred_rev_chg_q3"):.1f}B.')
print(f'Borrowing at 1.2× coverage against a {s["Tc"]:.0f}-yr $40M/MW contract would mean a loan of {dscr_adv:.0%} of the equipment cost.')
print(f'Renewal claim implies ${RENEW_PRICE.iloc[-1]:.2f} per GPU-hour (old book ${implied["H100"]:.2f} + ${RENEW.iloc[-1]:.2f}); '
      f'= ${per_mw(RENEW_PRICE.iloc[-1], "H100"):.0f}M per MW for H100-class; today H100 executed ${NOW["H100"]:.2f}, H200 ${NOW["H200"]:.2f}, A100 ${NOW["A100"]:.2f}.')
print(f'Years after contract at the median life, contract ends revised for the 2027-28 roll-offs: {built_after_revised:.1f} (was {OUT["crwv_yrs_after_contract_median_life"]:.1f}).')
print('Second-life cash worth today at ' + f'{DISC:.0%}: ' + ', '.join(f'{n} ${x:.1f}B' for n, x in SL_PV.items()) + f' (§7, per-dollar basis: base ${OUT["crwv_second_life_pv_base_bn"]:.1f}B)')
print(f'Holding-company bullets ${BULLETS.sum():.1f}B; second-life cash 2027-35, undiscounted: ' + ', '.join(f'{n} ${x:.1f}B' for n, x in SL_CASH.sum().items()))
fmt = {c: '{:.1f}' for c in FIN.columns} | {'borrowed (% of cost)': '{:.0%}', 'loan rate': '{:.2%}', 'contract (yrs)': '{:.0f}'}
display(QQ.style.format('{:.2f}').format('{:.0%}', subset=pd.IndexSlice[['adj. EBITDA margin', 'operating margin (GAAP)', 'depreciation ÷ revenue',
                                                                          'interest ÷ revenue', 'debt ÷ net equipment'], :]),
        FIN.T.style.format('{:.1f}').format('{:.0%}', subset=pd.IndexSlice[['borrowed (% of cost)'], :]).format('{:.2%}', subset=pd.IndexSlice[['loan rate'], :]),
        pd.concat([BULLETS.rename('holding-company bonds due ($B)'), SL_CASH.round(2).add_prefix('second-life cash, ')], axis=1).fillna(0).sort_index().style.format('{:.1f}'))
```

```python
fig, (ax, bx) = plt.subplots(1, 2, figsize=(14, 6.2), gridspec_kw={'width_ratios': [1, 1.15]})
cols = dict(zip(S8, [AXIS, RAMP[1], RAMP[2], EQUITY]))
for name, s in S8.items():
    t, cash, debt = mw_path(**s, P2=SECOND_MW['Glut'])
    eq = np.r_[-(1 - s['adv']) * K, cash - debt].cumsum()
    ax.plot(np.r_[0, t], eq, color=cols[name], lw=2, marker='o', ms=3, label=f"{name}: ${s['P']:.0f}M/MW, {s['Tc']:.0f}-yr contract, {s['adv']:.0%} borrowed")
    ax.plot([s['Tc']], [eq[int(s['Tc'])]], marker='|', ms=14, color=cols[name], mew=2)
w = FIN.loc['Worry case (100% borrowed)']
w_short, w_paid = w['after-contract shortfall, glut ($M/MW)'], w["shareholders' cash during contract ($M/MW)"]
ax.annotate(f'Worry case: years 4–5 fall short\nby ${w_short:.1f}M/MW, after ${w_paid:.0f}M/MW\npaid out during the contract', (4.5, 84), xytext=(5.1, 30), fontsize=7.8, color=EQUITY,
            arrowprops=dict(arrowstyle='-', color=EQUITY, lw=0.6))
ax.axhline(0, color=MUTED, lw=0.8)
ax.set_xlabel('Years since installation (| = contract ends)'); ax.set_ylabel("Cumulative cash to shareholders, $M per MW")
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.14), ncol=1, fontsize=7.5); ax.set_xlim(0, np.ceil(LIFE_I))
ax.set_title(f'One MW, second life at the glut price (${SECOND_MW["Glut"]:.0f}M/MW)', fontsize=11)

x = np.arange(len(YEARS))
bx.bar(x, BULLETS.reindex(YEARS).fillna(0), color=INK2, width=0.6, label=f'Holding-company bonds due (${BULLETS.sum():.0f}B, paid in one lump)')
for n, colr, ls in [('High', RAMP[2], '-'), ('Base', EQUITY, '-'), ('Glut', AXIS, '-')]:
    bx.plot(x, SL_CASH[n], color=colr, lw=2, ls=ls, marker='o', ms=3.5,
            label=f'Second-life cash, {n.lower()} (${SECOND_MW[n]:.0f}M/MW; ${SL_CASH[n].sum():.0f}B in total)')
bx.set_xticks(x, YEARS); bx.grid(axis='x', visible=False)
bx.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f'${y:g}B'))
bx.set_ylabel('$B a year'); bx.legend(loc='upper left', fontsize=7.5)
bx.set_title("The Jun 2026 fleet's second life vs. CoreWeave, Inc.'s own bonds", fontsize=11)
fig.suptitle(f"New contracts repay their loans within the contract. The old fleet's second life (${SL_CASH['Base'].sum():.0f}B base) is about what the holding company's bonds need in 2030–33 (${BULLETS.sum():.0f}B)", x=0.01, ha='left', fontweight='bold', fontsize=13)
source_note(fig, 'Source: CoreWeave Q3 2026 release and call (Nov 10 2026: contract prices per MW, loan advance rates and yields, 2027–28 roll-offs); DDTL 5.5 release (Aug 10 2026; '
                 '70% advance rate from a secondary source); Q2 2026 10-Q debt table and Sep 2026 convertible (data/crwv_debt.csv); SOFR (FRED). Second-life prices per MW: A100 2026 low and today, '
                 'H100 today (Ornn/GetDeploying), 90% utilized, minus Q3 fixed cash costs per MW. Fleet and lives from 10 §7. Pre-tax. Model: 10 §8.')
fig.tight_layout(rect=(0, 0.03, 1, 1))
fig.savefig(CHARTS / '10_q3_leverage.png', dpi=200, bbox_inches='tight')
plt.show()
```

**What this shows** (`charts/10_q3_leverage.png` and the tables above):

**1. Q3 was better, not a different business.**
- Revenue per MW rose from ~$8.2M to ~$9.2M a year. Adjusted EBITDA margin 62%, GAAP operating margin +6% (−2% in Q2). Interest fell to 23% of revenue.
- Debt rose ~$11B in one quarter, to $46.9B: 83% of net equipment, up from 76%.
- Management says the weighted cost of debt is 7.9%, below the ~9.5% we derive from interest ÷ debt (§4). Some of the new debt was drawn late in the quarter, so it didn't cost a full quarter's interest.
- Customers prepaid $2.1B in Q3, and management expects another ~$9B over the next three months. That's financing CoreWeave doesn't pay interest on.

**2. The new contracts are a different business per MW (left).** Each MW costs ~$26.5M of equipment and ~$3.5M a year to run.

| Contract | Coverage of loan payments | To shareholders during the contract, per MW |
|---|---|---|
| Old book (~$9M/MW, 5 yrs) | 1.2× | ~$5M |
| New long ($22.5M/MW, 4–5 yrs, 90% borrowed) | 3.2× | ~$66M |
| New short ($40M/MW, 3 yrs) | 5–8× | ~$89–95M |

- The old book's 1.2× is exactly the DDTL loans' minimum coverage. The old contracts were priced to cover the loan and little more, which is the razor's edge.
- A new long contract pays the shareholders' ~$2.65M per MW back in its first year. The CFO's "one to two years" payback is consistent with this.
- **So the money printer is no longer only the second life.** If these prices hold, new contracts make money during the contract.

**3. The worry (borrow against shortage prices, then get stuck with the debt in a glut) is mostly blocked by how lenders size the loans.**
- Lenders lend a share of the equipment's cost (70–100%), not a multiple of the contract's cash. Borrowing at the old book's 1.2× coverage against a 3-year, $40M/MW contract would mean a loan of ~290% of the equipment's cost. No lender offers that.
- So higher prices mean the loan is repaid faster, not that more is borrowed per MW.
- The worry does apply to **DDTL 5.5**, the one facility built for it: 3-year contracts against ~5-year debt.
  - At the 70% advance, ~$8M per MW (about a third of the cost) is still owed when the contract ends. Borrowing 100% leaves ~$12M (46%).
  - If the MW then rents at the glut price, years 4–5 fall short of the loan payments by ~$2–6M per MW.
  - That's small next to the ~$89–95M per MW the contract paid out.
- The danger isn't the mismatch itself. It's whether the contract-period cash is still there when the shortfall arrives. CoreWeave's plan is to spend it: ~$38B of capex in 2026 and about the same in 2027.
- DDTL 5.5 is $2.6B of $46.9B of debt. Lenders charged ~1 percentage point more than DDTL 5.0 and rated it below investment grade (Ba2/BB+). That gap is the market's price for exactly this risk.

**4. Who gets the old fleet's second life? Largely the holding company's bondholders (right).**
- CoreWeave, Inc. (the holding company) owes **~$21B in bonds that come due in one lump between 2030 and 2033**: unsecured notes, euro notes and three convertibles.
- The Jun 2026 fleet's second-life cash in those years, at the base price, is almost the same amount year by year: 2031 ~$7.5B against $7.1B due, and 2032 ~$8.0B against $7.5B.
- At the glut price the second life (~$15B in total) is less than the bonds.
- So the "free and clear" cash from the old fleet is, in effect, already promised. The razor's edge moved from the contract loans, which are repaid within the contracts, to the holding company.
- The convertibles ($10.8B) only turn into shares above ~$98–120 a share. At ~$49 (Nov 10 close) they're debt.

**5. Management's renewal plan assumes the shortage lasts to 2028.**
- ~300k GPUs come off contract in 2027–28, roughly all the Hopper and Ampere chips installed before 2025. Management expects ~$10B a year of extra revenue from renewing them.
- That works out to **~$5.15 per GPU-hour**, against an old-book ~$1.35. For H100-class chips that's ~$35M per MW, close to the new short-contract price.
- Today the H100 trades at ~$3.10 and the H200 at ~$6.30 (executed). The renewal plan sits between them, so it needs today's prices to last two more years.
- Our "high" case ($21M per MW, the H100 today) is below management's plan, and even it gives a second life of ~$64B, three times the holding company's bonds. **That's the money printer, and it depends on 2027–28 renewal prices.**

**6. Revisions to earlier sections.**
- **§7's value, revised upward.** §7 valued the second life per dollar of equipment, scaled from what an A100 earns per dollar it cost. Here it's valued per MW: what a MW of old chips rents for minus CoreWeave's own Q3 running costs per MW. Per MW is the better basis: rents are paid per chip, and CoreWeave's costs are per MW.
- With the 2027–28 roll-offs, the Jun 2026 fleet's second life is worth **~$16B** today at the base price (~$9B glut, ~$37B high, discounted at 12%). §7 had ~$9B. That's ~60% of the stock's ~$27B value (Nov 10) at the base price, and more than all of it at the high price.
- **Active power is IT power.** The CFO called it "critical IT load", so §6's per-MW comparisons need no ~1.3× cooling adjustment. §7's 2023 split now leaves out the cooling multiplier too.

**Caveats.**
- The 10-Q isn't filed yet, so the Q3 debt figure is the carrying amount from the release, while Q2's is principal.
- DDTL 5.5's 70% advance rate and its re-lease criteria aren't public.
- Contract prices per MW are management's statements, and we assume new MWs cost what the old book's did. Blackwell and Rubin systems may cost more per MW, which would lower coverage.
- Fixed cash costs come from one quarter.
- Pre-tax throughout. Refinancing the holding company's bonds is ignored. They could simply be rolled over, as long as the bond market stays open to CoreWeave.

## 9 · Putting it together: how does CoreWeave get off the razor's edge?

**Why ask.** §3–8 each looked at one piece. This section adds them up for the post in one chart: the pause test from 06, CoreWeave edition. **If CoreWeave signed nothing more after Q3, would what it has already signed, plus its GPUs' second life, carry the debt of CoreWeave, Inc.?**

**Method.** Each year, 2027–2033, in $B:
- **Cash that reaches CoreWeave, Inc.** after the GPU loans are paid. The GPU loans are borrowed by subsidiaries and repaid from their contracts first (§8).
  - **Old contracts:** the $103.7B backlog at Jun 30, timed as the 10-Q says (41% in 2 years, 39% in years 3–4, 20% after). They keep the old book's ~62% cash margin and 1.2× loan coverage (§8), so ~1/6 of their cash is left over.
  - **New contracts:** the ~$32B net added in Q3 (backlog $132.5B at Sep 30), assumed to be 5-year contracts starting 2027 at $22.5M/MW, with ~84% cash margin and ~3.2× coverage (§8). ~2/3 of their cash is left over.
  - **Second life:** the Jun 2026 fleet after its contracts end, at §8's base price (an A100's executed rent per MW today), from `SL_CASH`.
- **What CoreWeave, Inc. owes.**
  - Interest and principal on its own bonds and convertibles (`data/crwv_debt.csv`).
  - The equity it still has to put into capacity being built, which the CFO puts at under $5B through 2027.
- **The running balance:** Sep 30 cash ($4.1B) plus everything above, for three second-life prices (§8's glut, base, high).

What this leaves out, on purpose: anything signed after Q3, growth capex beyond what's signed, taxes, and refinancing. It's the "stop here" case, not a forecast.

```python
YRS9 = np.arange(2027, 2034)
def by_year(total, start, months):
    s = pd.Series(total / months, pd.date_range(start, periods=months, freq='MS'))
    return s.groupby(s.index.year).sum().reindex(YRS9, fill_value=0.0)

# cash in: what's left for CoreWeave, Inc. after each contract's loan is paid
REV_OLD = sum(by_year(v('crwv_rpo') * v(k), start, n) for k, start, n in
              [('crwv_rpo_0_24m', '2026-07-01', 24), ('crwv_rpo_25_48m', '2028-07-01', 24), ('crwv_rpo_49_78m', '2030-07-01', 30)])
NEW_BN = v('crwv_backlog_q3') - (v('crwv_rpo') - v('crwv_rev_q3'))      # Q3's net additions
REV_NEW = by_year(NEW_BN, '2027-01-01', 12 * YRS1)
left = lambda name: (1 - FIXED / S8[name]['P']) * (1 - 1 / FIN.loc[name, 'coverage during contract (×)'])   # $ left per $ of revenue
IN = pd.DataFrame({'Old contracts, after their loans': REV_OLD * left('Old book'),
                   'New contracts (Q3), after their loans': REV_NEW * left('New long'),
                   'Second life of the Jun 2026 fleet (base)': SL_CASH['Base'].reindex(YRS9, fill_value=0.0)})

# cash out: CoreWeave, Inc.'s own bonds and convertibles, plus the equity still owed into the build-out
HC = DEBT[DEBT.kind == 'holding-company bullet']
OUT9 = pd.DataFrame({'Interest on its bonds': [sum(r.amount_bn * r.rate for r in HC.itertuples() if r.maturity.year > y) +
                                               sum(r.amount_bn * r.rate * (r.maturity.day_of_year / 365) for r in HC.itertuples() if r.maturity.year == y) for y in YRS9],
                     'Bonds repaid': BULLETS.reindex(YRS9, fill_value=0.0).values,
                     'Equity into capacity being built': [v('crwv_parent_capital_thru_2027') if y == 2027 else 0.0 for y in YRS9]}, index=YRS9)

START = v('crwv_cash_q3')
other_in = IN.iloc[:, :2].sum(axis=1)
PATHS = pd.DataFrame({n: START + (other_in + SL_CASH[n].reindex(YRS9, fill_value=0.0) - OUT9.sum(axis=1)).cumsum() for n in ['Glut', 'Base', 'High']})
NET = IN.sum(axis=1) - OUT9.sum(axis=1)
PATH_OLD = START + (IN.iloc[:, 0] - OUT9.sum(axis=1)).cumsum()   # the razor's edge: old contracts only, no new ones, no second life
low_year = PATHS['Base'].idxmin()

OUT.update({'s9_cash_left_2027_33_bn': IN.sum().sum(), 'S9_owed_2027_33_bn': OUT9.sum().sum(),
            's9_end_balance_base_bn': PATHS['Base'].iloc[-1], 's9_end_balance_glut_bn': PATHS['Glut'].iloc[-1],
            's9_low_balance_base_bn': PATHS['Base'].min(), 's9_end_balance_old_only_bn': PATH_OLD.iloc[-1], 's9_new_contract_share_of_cash': IN.iloc[:, 1].sum() / IN.sum().sum()})
print(f'Q3 net new backlog ${NEW_BN:.1f}B. $ left per $ of revenue: old {left("Old book"):.0%}, new {left("New long"):.0%}.')
print(f'2027-33: cash left for CoreWeave, Inc. ${IN.sum().sum():.1f}B (old {IN.iloc[:,0].sum():.1f}, new {IN.iloc[:,1].sum():.1f}, second life {IN.iloc[:,2].sum():.1f}); '
      f'owed ${OUT9.sum().sum():.1f}B (interest {OUT9.iloc[:,0].sum():.1f}, bonds {OUT9.iloc[:,1].sum():.1f}, equity {OUT9.iloc[:,2].sum():.1f}).')
print('Running balance, end of year ($B):'); display(pd.concat([IN, -OUT9, NET.rename('net'), PATHS.add_prefix('balance, '), PATH_OLD.rename('balance, old contracts only')], axis=1).T.round(1))
```

**What this shows** (`charts/10_escape.png` and the table above):

**1. On the old contracts alone, CoreWeave, Inc. runs out of cash in 2031.**
- The contracts signed through June were priced to pay their GPU loans and little else (1.2× coverage, §8). Of each revenue dollar, ~11¢ is left for CoreWeave, Inc. after the loan.
- On the $103.7B backlog, that's ~$10B over 2027–33.
- Against it, CoreWeave, Inc. owes ~$32B over the same years: $21B of its own bonds and convertibles due 2030–33, ~$6B of interest, and the ~$5B the CFO says it still has to put into capacity being built.
- Starting from $4.1B of cash, it's ~−$18B by 2033. That's the razor's edge: the old book repays the GPU lenders and leaves CoreWeave, Inc.'s own lenders to be refinanced.

**2. Two things get it off the edge, and both are already in hand.**
- **The new contracts:** ~$20–25M per MW instead of ~$9M, so ~58¢ of each revenue dollar is left after the loan. Q3's ~$32B of net new backlog alone leaves ~$19B over five years.
- **The second life:** once a contract ends its GPUs are owned outright. Renting the Jun 2026 fleet at an A100's price per MW today adds ~$27B by 2033 (§8).
- Together, cash never falls below ~$4B (2027, when the build-out equity is due) and ends 2033 at **~$29B** at today's old-chip prices, **~$16B** at the 2026 glut price, and ~$65B if old chips rent like today's H100s.

**3. The catch: this is the "stop here" case.**
- It counts only what's signed. CoreWeave plans ~$38B of capex again in 2027. Signed at today's prices, that adds cash. Signed at old-book prices (if the shortage ends), it puts CoreWeave back on the edge, with more holding-company debt to fund each round's equity.
- It assumes the Q3 contracts are long-term at $22.5M/MW with 90% loans. Some are 3-year contracts on 5-year debt (DDTL 5.5, §8). Those pay more per MW but have to be re-let.
- No taxes, and old contracts keep the old book's ~62% cash margin.

## Findings, by section

**§1 · Rent** (`10_rent_history.png`)
- The H100 fell ~80% from its 2023 shortage price (~$8) to ~$1.50–2.00 in late 2025 / early 2026, then rose: now ~$2.50 spot, ~$3.10 executed, ~$3.70 on-demand list.
- From Jan 2024 to now, H100 rent is roughly flat (−5% to +10% a year, depending on the measure), not the −19% a year 08 §3 used. **Revises 08 §3**: on this history lenders clear the ~14% threshold. One year of shortage isn't a trend.
- Newer chips rent for more: B200 ~$9.40, H200 ~$6.30, A100 ~$1.05 (executed, 30-day median).

**§2 · New vs. used** (`10_new_vs_used.png`)
- A used H100 is quoted at ~75% of a new card. A used A100 at $4–17k depending on memory and source. No used B200 market yet.
- For a lender, a used card recovers ~60% of an H100 system's all-in cost before forced-sale discounts. Sources disagree by up to 2×.

**§3 · Payback** (`10_payback.png`)
- The reader's A100 pays back in ~0.8 years. A new H100 system at today's prices ~1.8 years (90% utilized), B200 under a year; an H100 bought at the 2023 peak took ~1.7 years at 90% utilization and hasn't paid back at 60%.
- CoreWeave as a whole takes ~5.5 years: its contracts work out to ~$1.35 (H100-class) to ~$2.10 (B200-class) per GPU-hour, below spot. Bondholders own the contract price, not the spot price.

**§4 · Margin vs. cost of debt** (`10_margins.png`)
- CoreWeave: −2% operating margin (59% before depreciation), interest 25% of revenue, borrowing at ~9.5%. Per $1 of equipment: ~18¢ cash, ~17¢ wear-out, ~1.5¢ left vs. ~6.4¢ interest.
- Equinix (~23%) and Digital Realty (~16%) borrow at ~3% against 20–40-year buildings. At Equinix's margin, CoreWeave's equipment would earn ~7% against 9.5% interest.

**§5 · Second life** (`10_second_life.png`)
- A ~6½-year-old A100 earns ~30¢ of revenue and ~19¢ of cash a year per dollar its system cost new. CoreWeave's equipment earns 31¢ and 18¢ under contract. Rent per original dollar falls ~27% per year of chip age (snapshot across four generations).
- With the same Q2 cash, an 8-year accounting life gives a ~+12% operating margin, and a 10-year life ~+20%.
- Return a year on $1 of CoreWeave equipment:
  - −3% if the chips die when their contracts end;
  - ~9.4% over an 8-year life with a second life at today's A100 executed rent and 90% utilization;
  - ~5% at 60% utilization or at the 2026 low;
  - ~16% at on-demand list prices.

  Against 9.5% interest: break-even, not rescue.
- Lenders (68¢ + interest per $1) are covered by year 6 in the base case, year 7 at the glut price, and never with burnout. Year 6 for 2025–26 GPUs is 2030–32, when the unsecured notes mature.

**§6 · Contract prices** (`10_contracts.png`)
- ~$64B of ~$89B of contracts with known amounts were signed Mar 2025–Apr 2026, as H100 prices fell from ~$2.30 to ~$1.50 an hour. Meta's $21B came at the bottom (Mar 2026), when a 36-month B200 reservation was listed at ~$2.79 an hour. It's ~$7.16 now.
- CoreWeave earns ~$9M per MW of power a year, roughly the long-term market price when it signed. New Q3 3–6-month contracts: ~$40M per MW. No disclosed discounts. Pricing is a target margin on cost; the concessions are in stock (OpenAI, ~$350M) and prepayment terms.
- Today's executed B200 price matches CoreWeave's revenue at ~20% utilization (H100: ~42%). But the hourly market is thin, the prices are from a shortage, and the take-or-pay contracts are what made the debt possible.
- Meta's contracts are fixed to Dec 2031 and Dec 2032. The evidence on old chips is good: H100s rebooked at ~95% of their original price, and an A100 contract runs to 2029.
- Shareholders get ~0.6¢ a year per $1 of equipment during the contract and everything after it. Their return a year is ~15% over a 10-year life at today's A100-like rents (~7.5% at the 2026 glut price), against −48% if the chips die with the contracts.

**§7 · Chip lives** (`10_chip_lives.png`)
- CoreWeave's 1.5 GW at Jun 2026 (estimated split): GB300 52%, GB200 22%, H100 15%, H200 8%, Ampere 3%.
- Azure/Google retired K80, P40, P100 and V100 at 7.0–9.5 years from launch (median 8.3). V100s are still listed by 17 providers at 9.4 years; an A100 is contracted to ~9 years.
- Retirement runs from launch. CoreWeave installs late and sells ~5-year contracts, so at the median life the fleet gets ~2.5 years after its contracts (~1.2 at 7 years, ~3.6 at 9.5).
- Second life of the Jun 2026 fleet, discounted at 12%: ~$9B (base; **revised to ~$16B in §8**), ~$5B (glut), ~$14B (10-year life), against a ~$27B market value (Nov 10). The cash per $1 is about what the contracts paid (18.7¢ vs 18.2¢); what changes is that none of it goes to lenders.

**§8 · After Q3** (`10_q3_leverage.png`)
- Q3: revenue per MW ~$9.2M (from ~$8.2M), adjusted EBITDA margin 62%, GAAP operating margin +6%, interest 23% of revenue. Debt +$11B to $46.9B (83% of net equipment). Prepayments +$2.1B, ~$9B more expected within 3 months.
- Per MW: new long contracts ($22.5M) cover loan payments 3.2× and pay ~$66M to shareholders over 5 years. The old book: 1.2× and ~$5M. Short contracts ($40M): 5–8×.
- Lenders size loans at 70–100% of cost, so higher prices don't mean more debt per MW. 1.2× coverage on a $40M contract would be a ~290% loan.
- DDTL 5.5 ($2.6B; 3-year contracts, 5-year debt) leaves ~$8–12M per MW owed at contract end. In a glut, it's short ~$2–6M per MW in years 4–5.
- Holding-company bonds due in single payments: ~$21B in 2030–33, matching the base-case second-life cash year by year. At the glut price (~$15B) the second life falls short.
- Management's 2027–28 renewal plan implies ~$5.15 per GPU-hour (~$35M per MW): today's shortage prices for two more years.
- Revises §7: per MW, and with the 2027–28 roll-offs, the second life is worth ~$16B (base), ~$9B (glut), ~$37B (high).

**§9 · Off the razor's edge** (`10_escape.png`)
- If CoreWeave signed nothing after Q3: the old contracts leave ~$10B for CoreWeave, Inc. over 2027–33 (11¢ per revenue dollar), against ~$32B it owes (bonds $21B, interest ~$6B, build-out equity ~$5B). Old book alone: −$18B by 2033.
- Q3's contracts (58¢ per dollar) add ~$19B and the second life ~$27B. Cash never falls below ~$4B and ends 2033 at ~$29B (base), ~$16B (glut), ~$65B (high).
- The escape holds only if the next rounds of capex are signed at prices like Q3's.

**Caveats**
- GPU prices come from aggregators with different methods (executed trades, list medians, marketplace medians). Levels differ by up to ~$1/hr for the same chip on the same day. Ornn's free history covers 3 months; GetDeploying's starts Nov 2025.
- Purchase and resale prices are dealer quotes and estimates. Nvidia publishes no list prices, and large buyers pay less.
- Per-GPU running costs are power and wholesale space only. Staff, sales, overhead, failures and spare capacity are excluded, so the per-GPU paybacks are optimistic.
- CoreWeave's company averages include capacity that isn't billing yet; the mature-contract economics are better (the Meta contract: ~3.3 years).
- §5 assumes a 2031 chip earns, per dollar it cost, what a 2026 A100 earns now. That price comes from a shortage year (the A100's spot price was $0.75 in May). Debt is treated as interest-only, and the returns are before tax.
- §7's fleet split is our estimate from public install dates, and its life range is 5 retirements at 2 hyperscalers.
- One quarter, annualized, for every margin. Digital Realty's latest quarter in SEC data is Q1 2026.
